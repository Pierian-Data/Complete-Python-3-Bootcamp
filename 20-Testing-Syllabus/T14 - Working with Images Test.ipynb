### Q4 - Multiple choice (1 point)

You want to rotate a square image by 90 degrees without losing any of its corners. Which call do you use?

A. `img.rotate(90)`
B. `img.rotate(90, expand=True)`
C. `img.resize((h, w))`
D. `img.convert('L').rotate(90)`

<details>
<summary>Solution</summary>

**Correct answer: B**

`rotate(90)` keeps the original canvas size and fills the empty corners, so a rotated rectangle gets its corners cut off. Passing `expand=True` grows the canvas so the whole rotated image fits.

**Why:** the course shows `pencils.rotate(90)` and `pencils.rotate(90, expand=True)` side by side and comments that at 120 degrees "the image is cut off".

</details>

### Q7 - Conceptual (1 point)

The course says a new colour can be obtained with `img.convert('L')`. What kind of picture does that produce, and what is the resulting `mode`? What does `convert('RGB')` do to an image that is already `'L'`?

Answer in words before you open the solution block below.

<details>
<summary>Solution</summary>

**Correct answer:** `convert('L')` converts the picture to greyscale: every pixel keeps only its brightness, and the resulting `mode` is `'L'`. `convert('RGB')` goes the other way - it turns a greyscale (or other single-band) image into a three-band full colour image with mode `'RGB'`, where the three colour channels all carry the same grey value.

**Why:** conversions never modify the original image in place; they return a new image, so you have to keep the result (`grey = img.convert('L')`) if you want to use it.

</details>

### Q8 - Conceptual (1 point)

In the course the blue image is saved with `blue.save("purple.png")` and the result is reopened with `Image.open("purple.png")`. How does Pillow know which file format to write, and what does `purple.format` report afterwards?

Answer in words before you open the solution block below.

<details>
<summary>Solution</summary>

**Correct answer:** Pillow picks the output format from the file extension of the name you give to `save()` - `.png` means PNG, `.jpg` means JPEG. When you reopen the file with `Image.open("purple.png")`, the `format` attribute reports `'PNG'`, and `format_description` gives the longer human readable description of the format.

**Why:** a freshly created image has `format` `None` because it never came from a file, so `format` is only meaningful after you have opened an existing image.

</details>

### Q14 - Coding exercise (2 points)

Add transparency with an alpha channel and save the result.

Create a fully opaque red image of size `(50, 50)` with `Image.new('RGBA', (50, 50), (255, 0, 0, 255))`, set its alpha to 128 with `putalpha(128)` and store it in a variable named **`transparent`**.

Create a blue image of size `(50, 50)` with `Image.new('RGBA', (50, 50), (0, 0, 255, 255))`, paste the transparent red on top of it with `blue.paste(transparent, box=(0, 0), mask=transparent)` and store the result in **`composite`**.

Finally save the composite into the scratch directory as `purple.png` (`os.path.join(TMPDIR, 'purple.png')`) and reopen it with `Image.open(...)` into a variable named **`reopened`**. Store the mode of the reopened image in **`reopened_mode``.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import os

from PIL import Image

transparent = Image.new('RGBA', (50, 50), (255, 0, 0, 255))
transparent.putalpha(128)

blue = Image.new('RGBA', (50, 50), (0, 0, 255, 255))
blue.paste(transparent, box=(0, 0), mask=transparent)
composite = blue

out_path = os.path.join(TMPDIR, 'purple.png')
composite.save(out_path)

reopened = Image.open(out_path)
reopened_mode = reopened.mode
```

**Why:** `putalpha(128)` makes the red half transparent, and using the image as the `mask` when pasting means only the transparent pixels of the red are blended over the blue - the 'purple' result from the course.

</details>

````python
assert transparent.mode == 'RGBA'
assert composite.size == (50, 50)
assert reopened_mode == 'RGBA'
assert reopened.size == (50, 50)
````
