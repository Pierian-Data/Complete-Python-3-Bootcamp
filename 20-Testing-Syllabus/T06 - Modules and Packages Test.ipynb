## Q1. Multiple choice

In Python, what is a module?

A) A Python file with the `.py` extension that implements a set of functions
B) A folder that contains Python files
C) A single function inside a Python file
D) A compiled `.pyc` file

<details>
<summary>Solution</summary>

**Correct answer: A.** A Python file with the `.py` extension that implements a set of

## Q5. Conceptual

What is `sys.path`, and how does Python use it when you write `import mymodule`?

<details>
<summary>Solution</summary>

**Answer:** `sys.path` is the list of directories that Python searches when it is asked to
import a module. Python walks that list from the front and uses the first directory
that contains a matching file (or a package folder with an `__init__.py`). The
directory of the script you are running is normally the first entry, which is why a
module sitting next to your script can be imported by name. You can inspect it with
`import sys; print(sys.path)` and change it with `sys.path.append(...)` or
`sys.path.insert(0, ...)`.

</details>

## Q6. Conceptual

What is the `if __name__ == "__main__":` guard for, and what is `__name__` equal to
when a file is run directly versus when it is imported?

<details>
<summary>Solution</summary>

**Answer:** `__name__` is a built-in variable that evaluates to the name of the current
module. When you run a script directly (`python one.py`) it is set to the string
`"__main__"`; when the file is imported by another module it is set to the module name
(for example `"one"`). The guard `if __name__ == "__main__":` therefore lets you write
code that only runs when the file is executed as a script. Without it, importing the
file would also run its top level code, which is exactly what `one.py` and `two.py`
demonstrate in the course folder.

</details>

## Q7. Conceptual

What does `from mymodule import *` do, and why is it considered risky?

<details>
<summary>Solution</summary>

**Answer:** It imports every public name (every name that does not start with an underscore)
from the module into the current namespace, so you can call `my_func()` directly
without the `mymodule.` prefix. The risk is that those names silently overwrite
names you already have in your own file - and that the reader can no longer see
where a name came from. Importing exactly what you need (`from mymodule import
my_func`) is clearer and safer.

</details>

## Q9. Coding (2 points)

Import the standard library modules with aliases: import `math` as `m` and import `os`
as `o`. Then define:

* `circle_area(radius)` returning `m.pi * radius ** 2`
* `list_names(path)` returning the sorted list of the names in `path` (use
  `o.listdir(path)`)

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import math as m
import os as o


def circle_area(radius):
    return m.pi * radius ** 2


def list_names(path):
    return sorted(o.listdir(path))
```

**Why:** An aliased import lets you write `m.pi` and `o.listdir(...)` instead of the longer `math.pi` and `os.listdir(...)`.

</details>

````python
import math
assert abs(circle_area(1) - math.pi) < 1e-9
assert abs(circle_area(2) - 4 * math.pi) < 1e-9
names = list_names('.')
assert isinstance(names, list)
assert all(isinstance(name, str) for name in names)
assert names == sorted(names)
````

## Q11. Coding (2 points)

Write the wildcard import `from math import *` at the **top level of the cell** (a
wildcard import is only allowed at module level, never inside a function). Then define:

* `wildcard_imported_names()` returning the sorted list of the public names (the names
  that do not start with an underscore) that are now in the module namespace

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from math import *


def wildcard_imported_names():
    return sorted(name for name in globals() if not name.startswith('_'))
```

**Why:** The wildcard import dumps every public name of the module into the namespace, which is exactly why it is risky: names such as `pi` and `sqrt` are now loose globals that can collide with your own names.

</details>

````python
names = wildcard_imported_names()
assert 'pi' in names
assert 'sqrt' in names
assert 'atan2' in names
assert all(not name.startswith('_') for name in names)
assert names == sorted(names)
````

## Q12. Coding (2 points)

Import `sys` and define:

* `add_search_dir(directory)` which inserts `directory` at the front of `sys.path`
  (`sys.path.insert(0, directory)`) and returns the new first entry of `sys.path`

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import sys


def add_search_dir(directory):
    sys.path.insert(0, directory)
    return sys.path[0]
```

**Why:** `sys.path` is the search path Python uses to find modules; inserting a directory at the front makes Python look there first.

</details>

````python
import sys
original = list(sys.path)
try:
    assert add_search_dir('C:/example/modules') == 'C:/example/modules'
    assert sys.path[0] == 'C:/example/modules'
finally:
    while 'C:/example/modules' in sys.path:
        sys.path.remove('C:/example/modules')
assert sys.path == original
````

## Q13. Coding (2 points)

Import `os` and define:

* `is_package_directory(directory)` returning `True` when the given directory contains
  an `__init__.py` file and `False` otherwise (use
  `os.path.join(directory, '__init__.py')` and `os.path.isfile(...)`)

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import os


def is_package_directory(directory):
    return os.path.isfile(os.path.join(directory, '__init__.py'))
```

**Why:** The `__init__.py` file is what tells Python that the folder is a package, so its presence is the test for a package directory.

</details>

````python
import os


def _course_root(start):
    current = os.path.abspath(start)
    while True:
        if os.path.isdir(os.path.join(current, '06-Modules and Packages')):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            return None
        current = parent


root = _course_root(os.getcwd())
assert root is not None, 'the course folder was not found'
modules_root = os.path.join(root, '06-Modules and Packages', '00-Modules_and_Packages')
assert is_package_directory(os.path.join(modules_root, 'MyMainPackage'))
assert is_package_directory(os.path.join(modules_root, 'MyMainPackage', 'SubPackage'))
assert not is_package_directory(modules_root)
````

## Q14. Coding (2 points)

Define:

* `fetch_attribute(module_name, attribute)` which imports the module called
  `module_name` and returns the attribute called `attribute` from it, so that
  `fetch_attribute('math', 'sqrt')` gives you the `math.sqrt` function

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def fetch_attribute(module_name, attribute):
    module = __import__(module_name, fromlist=[attribute])
    return getattr(module, attribute)
```

**Why:** `__import__` with a `fromlist` behaves like `from module import attribute`, and `getattr` pulls the named object out of the imported module.

</details>

````python
import math
import os
assert fetch_attribute('math', 'pi') == math.pi
assert fetch_attribute('math', 'sqrt')(9) == 3.0
assert fetch_attribute('os', 'getcwd')() == os.getcwd()
````
