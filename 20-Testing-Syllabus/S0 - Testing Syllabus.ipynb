# Testing Syllabus - Complete Python 3 Bootcamp

A self-assessment suite for this course repository. It turns every lecture
notebook into a marked test, so you can measure whether the material actually
stuck before you move on.

**What you get**

- 20 module tests (`T00` ... `T19`), one per course section, 20 points each.
- 1 cumulative final exam (`99`), 36 questions, 60 points.
- A blueprint below: what is tested, how much it is worth, what to revise when
  you fail, and the order the modules build on each other.

**How it is marked**

Every coding exercise has a `check` cell containing `assert` statements. The
grader replays the notebook and reports each check as PASS or FAIL, so you get
an instant, objective score for the coding half of every test. The prose
questions (multiple choice, conceptual, predict-the-output) are marked against
the collapsed solution next to them.

**Total course mark**

```
section tests  : 400 points  ->  85% of the final grade
final exam     :  60 points  ->  15% of the final grade
```

## 5. Scoring model and pass bands

**Per test (20 points)**

| Score | Result |
| --- | --- |
| 18 - 20 | Mastered. Move on. |
| 14 - 17 | Pass (70%). Re-read the flagged lecture topics, then retake. |
| 10 - 13 | Weak. Redo the lecture notebooks for that section before retaking. |
| 0 - 9 | Not yet covered. Work through the section from the start. |

**Coding checks specifically.** Each check cell is worth the 2 points of its
exercise. A failing check means the identifier, the logic, or the return value
is wrong - the grader prints the failing assertion so you know which one.

**Final course grade**

```
course grade = 0.85 * (weighted section points / 400) + 0.15 * (final exam / 60)
```

| Course grade | Band |
| --- | --- |
| 85 - 100 | Distinction |
| 70 - 84 | Merit |
| 55 - 69 | Pass |
| below 55 | Not yet at standard |

**Final exam gate:** 42 / 60 (70%) is a pass. The exam is the only test that
covers sections you never touched, so treat it as the real measure.

**Gate rule.** A module is "signed off" at 14 / 20. Sign off on T00 - T03
before starting OOP, and on T03 before T05, T07 and T10.

One pass through the course material plus the matching test. Move on only when
the test is signed off (14 / 20).

| 1 | Sections 00 - 01 | T00, T01 | T00 >= 16, T01 >= 14 |
| 2 | Section 02 | T02 | >= 16 |
| 3 | Section 03 | T03 | >= 16 |
| 4 | Section 04, start 05 | T04, T05 | >= 14 each |
| 5 | Sections 06 - 07 | T06, T07 | >= 14 each |
| 6 | Sections 09 - 12 | T09, T10, T11, T12 | >= 14 each |
| 7 | Sections 13 - 17, 19 | T13 - T17, T19 | >= 14 each |
| 8 | Sections 08, 18, full revision | T08, T18, 99 | exam >= 42 / 60 |

Weeks 4 - 5 are the ones people rush. OOP and exceptions are the two sections
that gate the most later material: T05 and T07 are prerequisites for T08, T10,
T12, T15, T16 and T18.

**If you are short on time:** T00, T02, T03, T05, T07 and T17 carry 50% of the
section weight. Do those thoroughly and skim the rest.

## 8. Your mark sheet and your portfolio

Two more artifacts sit alongside the tests.

**`S2 - Grading Records.ipynb` - the mark sheet.** Every attempt is logged in
`records.json`; the notebook renders it as per-module best scores, the weighted
course mark, the band, and the next module to attack. Log an attempt from the
command line right after you finish a test:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus/grade_records.py run "T03 - Methods and Functions Test.ipynb" --prose 7
```

**`S3 - Project Portfolio.ipynb` - the portfolio.** One project per course
section, with the deliverables to tick off and the README entry to copy into
your own repository. The capstone (T18) is the flagship; the other nineteen
are supporting evidence.

The two belong together: the portfolio shows what you built, the grading
record shows the work behind it. `grade_records.py report` exports the record
as `GRADE_REPORT.md`, ready to commit into a portfolio repository.

**And there is a professor.** `PROFESSOR.md` at the repository root is the
operating manual for an AI tutor that teaches this course: it diagnoses the
misconception before explaining, quotes the actual lecture notebook, verifies
with the grader, and logs the attempt. `AGENTS.md` makes any coding agent in
this repository behave that way, and `professor.py` in this folder is a local
tutor CLI you can talk to from a terminal. See `S4 - AI Tutor.ipynb`.

**And the record is tamper-evident.** The ledger in `records.json` is a hash
chain: every attempt carries the hash of its predecessor and of its own
content, plus the SHA-256 of the notebook that was graded. A signed
`SEAL.json` binds the chain head to a manifest of every file in the suite, so
neither a grade nor a weakened test can be changed quietly. Redoing a test is
documented, never erased: the redo is a new entry that supersedes the old one.
`INTEGRITY.md` explains the scheme, its limits, and how to anchor the seal
outside this repository.

## 9. Retake policy and how to use the grader

1. Work through the course notebooks for the section.
2. Open the matching `T**` notebook. Read the question, write your answer in
   the code cell, run it.
3. Run the grader to get an objective score on the coding half:

   ```
   C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade.py "T03 - Methods and Functions Test.ipynb"
   ```

   It prints one line per check: PASS or FAIL with the failing assertion.
   Without arguments it grades every test in the folder.

4. Mark the prose questions yourself against the collapsed solutions.
5. Re-read the course notebook for any topic you missed, then retake.
6. Log the attempt so it counts towards your mark sheet:

   ```
   C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus/grade_records.py run "T03 - Methods and Functions Test.ipynb" --prose 7
   ```

There is no penalty for retaking. The score you submit is the one from your
final attempt, and the ledger keeps the best result per module.

**Auto-marked coverage.** The 6 coding exercises per test are the objective
half; on a standard test that is 12 of the 20 points. On the project modules
it is 4 of 20, because the rubric questions carry the rest.

## 10. Projects, the portfolio and the progress log

The tests prove you can *use* a feature. The projects prove you can *build*
with it. They are graded differently, on purpose.

**Project grading - a shared rubric.** `rubrics.json` holds one rubric used
for every project in the course, so the marks are comparable:

| Criterion | Weight | Asks |
| --- | --- | --- |
| Functionality | 3 | Does it do what the brief asks? |
| Robustness | 2 | Does it survive bad input and edge cases? |
| Tests | 2 | Is there a suite, does it pass, does it cover the core? |
| Code quality and structure | 2 | Clear names, sensible split, no dead code? |
| Documentation | 2 | A README that says what it is and how to run it? |
| Use of the section's skills | 1 | Does it use what the section taught? |

Each criterion is scored 0-4: 0 not attempted, 1 partial, 2 basic, 3 solid,
4 excellent. The mark is `sum(score x weight) / (4 x total weight)`. Project
marks are reported next to the course mark, not folded into it.

```bash
python 20-Testing-Syllabus/grade_records.py rubric P05
python 20-Testing-Syllabus/grade_records.py record-project P05 --scores 3,4,2,3,4,3
```

**The portfolio** (`S3 - Project Portfolio.ipynb`) is where the projects are
specified: one per section, with the deliverables and the README entry to
copy. The capstone (T18) is the flagship; the other nineteen are evidence.

**The progress log.** A dated, chained journal of what you actually did:

```bash
python 20-Testing-Syllabus/grade_records.py log "rebuilt the OOP model, still stuck on super()"
```

**The written reports.** The professor writes them from your record:

```bash
python 20-Testing-Syllabus/professor.py review --project P05 --path ../my-project
python 20-Testing-Syllabus/professor.py progress
```

The first reviews one project against its rubric; the second analyses your
whole record and says what to work on next. Both are saved to `reviews/` and
both are logged, so the advice is part of the history.
