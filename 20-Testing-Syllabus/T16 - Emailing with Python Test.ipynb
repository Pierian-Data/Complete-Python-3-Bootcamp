### Q2 - Multiple choice (1 point)

You connect to Gmail with `smtplib.SMTP('smtp.gmail.com', 587)` and then call `smtp_object.starttls()`. Why is `starttls()` needed on port 587?

A. It compresses the message so it arrives faster
B. It upgrades the connection to encrypted TLS
C. It requests a larger window from the server
D. It tells the server to retry the connection

<details>
<summary>Solution</summary>

**Correct answer: B**

</details>

import smtplib

from email.message import EmailMessage


def send_report(password):
    message = EmailMessage()
    message['Subject'] = 'Weekly Report'
    message['From'] = 'sender@example.com'
    message['To'] = 'receiver@example.com'
    message.set_content('Here is the report.')

    smtp_object.ehlo()

Read a received email that has been fetched as a raw string.

Take this raw email:

```text
From: sender@example.com
To: receiver@example.com
Subject: Test Email
Content-Type: text/plain; charset="utf-8"

Hello from the inbox
```

Store it in a variable named **`raw_email`**, parse it with `email.message_from_string(raw_email)` into a variable named **`parsed`**, and store:

- **`subject`** - `parsed['Subject']`
- **`from_address`** - `parsed['From']`
- **`body`** - the decoded text of the plain text part, found by walking the parts, checking `part.get_content_type() == 'text/plain'` and using `part.get_payload(decode=True).decode('utf-8')`

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import email

raw_email = ('From: sender@example.com\n'
             'To: receiver@example.com\n'
             'Subject: Test Email\n'
             'Content-Type: text/plain; charset="utf-8"\n'
             '\n'
             'Hello from the inbox')

parsed = email.message_from_string(raw_email)

subject = parsed['Subject']
from_address = parsed['From']

for part in parsed.walk():
    if part.get_content_type() == 'text/plain':
        body = part.get_payload(decode=True).decode('utf-8')
```

**Why:** This is the same walk over the MIME parts that the course performs on a mail fetched with `M.fetch(data[0], '(RFC822)')`.

</details>

````python
assert subject == 'Test Email'
assert from_address == 'sender@example.com'
assert body == 'Hello from the inbox'
````

### Q14 - Coding exercise (2 points)

Read a *multipart* received email and pick out the pieces.

Take this raw email:

```text
From: sender@example.com
To: receiver@example.com
Subject: Report
MIME-Version: 1.0
Content-Type: multipart/alternative; boundary="BOUND"

--BOUND
Content-Type: text/plain; charset="utf-8"

Plain version

<p>HTML version</p>

```

Store it in **`raw_email`**, parse it with `email.message_from_string(raw_email)` into **`parsed`**, and store:

- **`subject`** - `parsed['Subject']`
- **`part_types`** - a list of the content types of every part, from `parsed.walk()`
- **`plain_body`** - the decoded text of the `text/plain` part
- **`html_body`** - the decoded text of the `text/html` part

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import email

raw_email = ('From: sender@example.com\n'
             'To: receiver@example.com\n'
             'Subject: Report\n'
             'MIME-Version: 1.0\n'
             'Content-Type: multipart/alternative; boundary="BOUND"\n'
             '\n'
             '--BOUND\n'
             'Content-Type: text/plain; charset="utf-8"\n'
             '\n'
             'Plain version\n'
             '--BOUND\n'
             'Content-Type: text/html; charset="utf-8"\n'
             '\n'
             '<p>HTML version</p>\n'
             '--BOUND--\n')

parsed = email.message_from_string(raw_email)

subject = parsed['Subject']
part_types = [part.get_content_type() for part in parsed.walk()]

for part in parsed.walk():
    if part.get_content_type() == 'text/plain':
        plain_body = part.get_payload(decode=True).decode('utf-8')
    elif part.get_content_type() == 'text/html':
        html_body = part.get_payload(decode=True).decode('utf-8')
```

**Why:** A multipart message has one container part plus one part per body, so `walk()` gives you the container first and then the bodies it holds.

</details>

````python
assert subject == 'Report'
assert part_types == ['multipart/alternative', 'text/plain', 'text/html']
assert plain_body == 'Plain version'
assert html_body == '<p>HTML version</p>'
````
