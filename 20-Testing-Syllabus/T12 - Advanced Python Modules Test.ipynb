## Q1. Multiple choice

What happens when you index a `defaultdict` with a key that does not exist yet?

A. A `KeyError` is raised, exactly like a plain `dict`
B. The missing key is created and filled with the value produced by the `default_factory`
C. `None` is returned
D. The whole dictionary is cleared

<details>
<summary>Solution</summary>

**Answer: B** - the missing key is created and filled with the value produced by the
`default_factory`.

**Why:** a `defaultdict` never raises `KeyError`. `defaultdict(list)` creates an empty list for a
new key, `defaultdict(lambda: 0)` creates a zero, and `defaultdict(object)` creates a new empty
object.

</details>

## Q3. Multiple choice

Which regular expression call returns **every** match in the text as a list?

A. `re.match(pattern, text)`
B. `re.search(pattern, text)`
C. `re.findall(pattern, text)`
D. `re.compile(pattern)`

<details>
<summary>Solution</summary>

**Answer: C** - `re.findall(pattern, text)`.

**Why:** `re.search()` returns only the first match (a Match object, or `None` if there is none),
`re.match()` only tries to match at the very beginning of the string, and `re.compile()` only
prepares a pattern for repeated use without searching anything.

</details>

## Q7. Conceptual - timing code with `timeit`

Consider:

```text
import timeit

setup = '''
def func_one(n):
    return [str(num) for num in range(n)]
'''

stmt = 'func_one(100)'

elapsed = timeit.timeit(stmt, setup, number=1000)
```

What does `elapsed` contain, and what is the difference between this and the `time.time()` approach
shown earlier in the same notebook?

<details>
<summary>Solution</summary>

**What `elapsed` contains:** the total number of seconds it took to run `stmt` a
thousand times (a float, so it is the average per run multiplied by 1000).

**Difference from `time.time()`:** the `time.time()` approach runs the code once and measures it
with the clock, and because of the precision of the clock the code has to take at least about
0.1 seconds for the measurement to be meaningful. `timeit` runs the statement many times and
reports the total, which makes it much better for very fast blocks of code. The `setup` string
holds anything that must be defined first, such as the function definitions.

</details>

Name the two `zipfile` methods used for this.

<summary>Solution</summary>

## Q9. Coding - counting with `Counter`

a list of `(element, count)` pairs.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from collections import Counter

def count_letters(text):
    return Counter(text)
```

**Why:** `Counter` is a `dict` subclass that counts hashable elements, so it can be built straight from the string.

</details>

## Q10. Coding - grouping with `defaultdict`

Write a function `group_by_initial(words)` that returns a `collections.defaultdict` mapping the
first letter of each word to the list of words that start with that letter.

The returned object must be a `defaultdict` with `list` as its default factory, so asking for a
letter that does not occur returns an empty list instead of raising `KeyError`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from collections import defaultdict

def group_by_initial(words):
    groups = defaultdict(list)
    for word in words:
        groups[word[0]].append(word)
    return groups
```

**Why:** `defaultdict(list)` supplies an empty list for every new key, so the loop can append straight away.

</details>

````python
from collections import defaultdict

groups = group_by_initial(['apple', 'avocado', 'banana', 'blueberry', 'cherry'])
assert isinstance(groups, defaultdict)
assert groups['a'] == ['apple', 'avocado']
assert groups['b'] == ['banana', 'blueberry']
````

## Q11. Coding - a `namedtuple`

Use `collections.namedtuple` to create a record type named `Dog` with the fields `age`, `breed`
and `name` (in that order).

Then write a function `make_dog(age, breed, name)` that returns a `Dog` instance with those
values. The fields must be reachable both as attributes (`dog.age`) and by index (`dog[0]`).

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from collections import namedtuple

Dog = namedtuple('Dog', ['age', 'breed', 'name'])

def make_dog(age, breed, name):
    return Dog(age=age, breed=breed, name=name)
```

**Why:** `namedtuple('Dog', [...])` creates a new class whose instances are tuples with named fields.

## Q12. Coding - working with folders using `os`

Write a function `find_files(directory, extension)` that returns a **sorted** list of the names
of the files directly inside `directory` whose name ends with `extension`.

Use `os.listdir()` to read the directory and `os.path` for the path handling.

For example, for a folder holding `notes.txt`, `report.pdf` and `data.txt`, calling
`find_files(folder, '.txt')` should return `['data.txt', 'notes.txt']`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import os

def find_files(directory, extension):
    return sorted(name for name in os.listdir(directory) if name.endswith(extension))
```

**Why:** `os.listdir()` returns the names of the entries in the directory; `sorted()` puts them in a predictable order.

</details>

````python
import os
import shutil
import tempfile

tmp_dir = tempfile.mkdtemp()
try:
    open(os.path.join(tmp_dir, 'notes.txt'), 'w').write('hello')
    open(os.path.join(tmp_dir, 'report.pdf'), 'w').write('hello')
    open(os.path.join(tmp_dir, 'data.txt'), 'w').write('hello')

    assert find_files(tmp_dir, '.txt') == ['data.txt', 'notes.txt']
    assert find_files(tmp_dir, '.pdf') == ['report.pdf']
    assert find_files(tmp_dir, '.csv') == []
finally:
    shutil.rmtree(tmp_dir, ignore_errors=True)
````

## Q13. Coding - dates with `datetime`

Write two functions:

- `days_between(d1, d2)` - takes two `datetime.date` objects and returns the number of whole
  days from `d1` to `d2` (as an `int`).
- `format_date(d)` - takes a `datetime.date` object and returns it as a string in the format
  `YYYY-MM-DD` (for example `2015-03-11`).

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from datetime import date

def days_between(d1, d2):
    return (d2 - d1).days

def format_date(d):
    return d.strftime('%Y-%m-%d')
```

**Why:** Subtracting two dates gives a `timedelta`, whose `.days` attribute is the number of whole days; `strftime` formats a date with the given pattern.

</details>

````python
import datetime

d1 = datetime.date(2015, 3, 11)
d2 = datetime.date(1990, 3, 11)
assert days_between(d2, d1) == 9131
assert days_between(datetime.date(2020, 1, 1), datetime.date(2020, 1, 31)) == 30
assert format_date(d1) == '2015-03-11'
````

## Q14. Coding - searching text with regular expressions

Write a function `find_phone_numbers(text)` that returns a list of every phone number of the
form `###-###-####` found in `text`, using `re.findall` with the pattern `r'\d{3}-\d{3}-\d{4}'`.

For example, for the text `"call 408-555-1234 or 212-555-0000"` the result should be
`['408-555-1234', '212-555-0000']`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import re

def find_phone_numbers(text):
    return re.findall(r'\d{3}-\d{3}-\d{4}', text)
```

**Why:** `re.findall` returns every match of the pattern as a list of strings; `\d{3}` matches exactly three digits.

</details>

````python
text = "call 408-555-1234 or 212-555-0000"
assert find_phone_numbers(text) == ['408-555-1234', '212-555-0000']
assert find_phone_numbers("no numbers here") == []
assert find_phone_numbers("123-45-6789 and 123-456-7890") == ['123-456-7890']
````
