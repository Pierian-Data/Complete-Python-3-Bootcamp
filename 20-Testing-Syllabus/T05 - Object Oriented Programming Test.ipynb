## Q1. Multiple choice

Which Python keyword is used to define a new user defined object type?

A) `def`
B) `class`
C) `object`
D) `self`

<details>
<summary>Solution</summary>

**Correct answer: B.** `class`. The `class` keyword creates a blueprint for a new
object type; instances are made by calling the class, for example `x = Sample()`.
`def` defines functions, `object` is the base class of everything in Python and `self`
is the name given by convention to the first parameter of an instance method.

</details>

## Q2. Multiple choice

Inside a class, what is the name given by convention to the first parameter of every
instance method?

A) `this`
B) `cls`
C) `self`
D) `me`

<details>
<summary>Solution</summary>

**Correct answer: C.** `self`. Each attribute assignment in a method starts with a
reference to the instance object itself, and by convention that reference is named
`self`, as in `def __init__(self, breed): self.breed = breed`.

</details>

## Q3. Multiple choice

Where is a "class object attribute" (a shared attribute such as `species = 'mammal'`)
defined?

A) Inside `__init__`, using `self.`
B) Inside any method, using `self.`
C) Directly in the class body, outside of any method
D) Only after the class definition, with `ClassName.attribute = ...`

<details>
<summary>Solution</summary>

**Correct answer: C.** Directly in the class body, outside of any method. Class object
attributes are shared by every instance of the class and are placed in the class body
before `__init__` by convention. Attributes written with `self.` belong to one
instance only.

</details>

## Q6. Conceptual

When exactly is the `__init__` method of a class called, and what is the effect of
writing `self.breed = breed` inside it?

<details>
<summary>Solution</summary>

**Answer:** `__init__` is called automatically right after the object has been created,
that is at instantiation time (`sam = Dog(breed='Lab')`); it is the constructor of the
class. `self.breed = breed` stores the value that was passed in as the `breed`
argument on the new instance, so every instance gets its own `breed` attribute.
`self` is the reference to the instance object itself.

</details>

## Q8. Conceptual

In the inheritance example `Dog` derives from `Animal`. Explain the difference between
the three things a derived class can do with the base class methods, and say what
`super()` is used for.

<details>
<summary>Solution</summary>

**Answer:** A derived class can inherit the functionality of the base class (for example
`Dog` inherits `eat()` from `Animal`), override existing behaviour (for example
`Dog.whoAmI()` replaces `Animal.whoAmI()`), and extend the functionality by defining
new methods (for example `Dog.bark()`). `super()` is used inside a derived class to
call the base class implementation explicitly, for example `super().__init__(name)`
runs `Animal.__init__` so the base class can set up its own attributes without
repeating that code in the subclass.

</details>

Create a class named **`Circle`** with:

* a method `setRadius(self, new_radius)` that updates `self.radius` and recomputes

* a method `getCircumference(self)` that returns `self.radius * self.pi * 2`

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class Circle:
    pi = 3.14

    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius * radius * Circle.pi

    def setRadius(self, new_radius):
        self.radius = new_radius
        self.area = new_radius * new_radius * self.pi

    def getCircumference(self):
        return self.radius * self.pi * 2
```

**Why:** `Circle.pi` is needed in `__init__` because the instance has no `pi` attribute yet.

</details>

````python
c = Circle()
assert c.radius == 1
assert abs(c.area - 3.14) < 1e-9
assert abs(c.getCircumference() - 6.28) < 1e-9
c.setRadius(2)
assert c.radius == 2
assert abs(c.area - 12.56) < 1e-9
assert abs(c.getCircumference() - 12.56) < 1e-9
````

Create two classes:

* **`Dog`** that inherits from `Animal`, calls `super().__init__(name)` from its
  `__init__(self, name)`, and overrides `speak(self)` to return `'<name> says Woof!'`

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement abstract method")


class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

    def speak(self):
        return self.name + ' says Woof!'
```

**Why:** `super()` calls the base constructor and the base `speak()` raises `NotImplementedError`.

</details>

````python
fido = Dog('Fido')
assert isinstance(fido, Animal)
assert fido.name == 'Fido'
assert fido.speak() == 'Fido says Woof!'
try:
    Animal('Generic').speak()
except NotImplementedError:
    pass
else:
    raise AssertionError("Animal.speak() should raise NotImplementedError")
````

## Q13. Coding (2 points)

Create a class named **`Vector`** with:

* `__init__(self, x, y)` storing `self.x` and `self.y`
* `__repr__(self)` returning `'Vector(<x>, <y>)'`, for example `'Vector(1, 2)'`
* `__add__(self, other)` returning a new `Vector` with the components added
* `__sub__(self, other)` returning a new `Vector` with the components subtracted
* `__mul__(self, scalar)` returning a new `Vector` with both components multiplied
  by `scalar`

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return "Vector(%s, %s)" % (self.x, self.y)

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)
```

**Why:** Dunder methods let the Python operators work on your own objects.

</details>

````python
v = Vector(1, 2)
w = Vector(3, 4)
assert (v + w).x == 4 and (v + w).y == 6
assert (w - v).x == 2 and (w - v).y == 2
assert (v * 3).x == 3 and (v * 3).y == 6
assert repr(v) == 'Vector(1, 2)'
````

## Q14. Coding (2 points)

Solve the OOP Challenge. Create a class named **`Account`** with:

* `__init__(self, owner, balance=0)` storing `self.owner` and `self.balance`
* `deposit(self, amount)` that adds `amount` to `self.balance` and returns the new
  balance; it must raise `ValueError` when `amount` is not positive
* `withdraw(self, amount)` that subtracts `amount` from `self.balance` and returns the
  new balance; it must raise `ValueError` when `amount` is bigger than the current
  balance, and the balance must stay unchanged in that case

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount
        return self.balance
```

**Why:** The balance is only changed after the amount has been validated, so an overdraft leaves the balance untouched.

</details>

````python
acct = Account('Jose', 100)
assert acct.owner == 'Jose'
assert acct.deposit(50) == 150
assert acct.balance == 150
assert acct.withdraw(75) == 75
assert acct.balance == 75
try:
    acct.withdraw(500)
except ValueError:
    pass
else:
    raise AssertionError("an overdraft should raise ValueError")
assert acct.balance == 75
````
