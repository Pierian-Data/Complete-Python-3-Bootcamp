## Q1. Multiple choice

In Python 3, what do `math.floor(4.35)` and `math.ceil(4.35)` return, and how do they differ from `round(4.35)`?

A. `math.floor` returns `4.35` and `math.ceil` returns `5.0`, so both keep the fractional part
B. `math.floor` returns the integer `4` and `math.ceil` returns the integer `5`, while `round(4.35)` returns the integer `4`
C. `math.floor` returns `5` and `math.ceil` returns `4`, so they are swapped
D. Both of them return the integer `4`, identical to `round(4.35)`

<details>
<summary>Solution</summary>

**Correct answer: B**

`math.floor(4.35)` returns `4` and `math.ceil(4.35)` returns `5`. Both always return an `int`, never a `float`: `floor` truncates toward negative infinity and `ceil` rounds toward positive infinity. `round(4.35)` also returns an `int` (`4`) because no digits argument was given; with a digits argument such as `round(4.35, 1)` it returns a `float` (`4.3`).

Key difference: `round` rounds to the *nearest* value, while `floor` and `ceil` snap to the surrounding integers. For negative numbers they clearly differ: `round(-4.35)` is `-4`, `math.floor(-4.35)` is `-5`, and `math.ceil(-4.35)` is `-4`.

</details>

## Q3. Multiple choice

What does `hex(255)` return?

A. The string `'0xff'`
B. The string `'255'`
C. The string `'0b11111111'`
D. The string `'ff'`

<details>
<summary>Solution</summary>

**Correct answer: A**

`hex(255)` returns the string `'0xff'`. The conversion functions always return strings with a prefix that tells you the base: `hex()` produces `0x`, `bin()` produces `0b`, and `oct()` produces `0o`. So `bin(255)` is `'0b11111111'` (option C is `bin`, not `hex`) and `oct(255)` would be `'0o377'`. Option D would be the raw digits without the `0x` prefix, which `hex()` never produces.

</details>

## Q4. Multiple choice

Which string method returns a copy of the string padded with a fill character so that the result is left-justified within the given width?

A. `ljust(width, fill)`
B. `rjust(width, fill)`
C. `center(width, fill)`
D. `zfill(width)`

<details>
<summary>Solution</summary>

**Correct answer: A**

`ljust(width, fill)` pads on the right, producing a left-justified string. `rjust` pads on the left (right-justified), `center` splits the padding between both sides, and `zfill` pads on the left with zeros only and takes no fill argument. All of these return a *new* string because strings are immutable.

</details>

## Q7. Conceptual - predict the output

Without running it, predict what this prints:

```
x = [3, 1, 2]
y = x.sort()
print(x, y)
```

<details>
<summary>Solution</summary>

**Output:**

```
[1, 2, 3] None
```

`list.sort()` sorts the list **in place** and returns `None`, so `y` is `None` and `x` itself is now sorted. This is the classic mistake the course warns about: you must not write `y = x.sort()` expecting a sorted copy. When you want a new sorted list and want to keep the original, use the built-in `sorted(x)`, which returns a new list and leaves `x` untouched.

</details>


```python
import math
```

## Q12. Coding

Write a function named **`analyze_words(words)`** that takes a list of words and returns a dictionary with exactly these keys:

- `'counts'` - a dictionary mapping each distinct word to the number of times it appears (use `collections.Counter`)
- `'by_length'` - a dictionary mapping each word length to the list of words of that length, in their original order, repeated words included (use `collections.defaultdict(list)`)
- `'longest'` - the longest word; if several words tie for longest, return the first one encountered
- `'first_letter_counts'` - a dictionary mapping each first letter to how many words start with that letter (use `dict` / `setdefault` rather than hard-coding it)
- `'ordered_counts'` - the same word/count pairs as `'counts'` but with the words in alphabetical order (use `collections.OrderedDict`)

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from collections import Counter, defaultdict, OrderedDict


def analyze_words(words):
    counts = Counter(words)

    by_length = defaultdict(list)
    for word in words:
        by_length[len(word)].append(word)

    first_letter_counts = {}
    for word in counts:
        letter = word[0]
        first_letter_counts[letter] = first_letter_counts.get(letter, 0) + counts[word]

    ordered_counts = OrderedDict(
        (word, counts[word]) for word in sorted(counts)
    )

    return {
        "counts": dict(counts),
        "by_length": dict(by_length),
        "longest": max(words, key=len),
        "first_letter_counts": first_letter_counts,
        "ordered_counts": dict(ordered_counts),
    }
```

**Why:** `Counter` counts hashable items, `defaultdict(list)` groups words by length without a membership test, `max(words, key=len)` keeps the first of any tie, and `OrderedDict` preserves the alphabetical insertion order.

</details>

````python
result = analyze_words(["a", "bb", "a", "ccc"])
assert result["counts"] == {"a": 2, "bb": 1, "ccc": 1}
assert result["by_length"] == {1: ["a", "a"], 2: ["bb"], 3: ["ccc"]}
assert result["longest"] == "ccc"
assert result["first_letter_counts"] == {"a": 2, "b": 1, "c": 1}
assert result["ordered_counts"] == {"a": 2, "bb": 1, "ccc": 1}
````

## Q13. Coding

Write a function named **`index_and_pair(names, scores)`** that takes two equal-length lists and returns a dictionary with exactly these keys:

- `'indexed'` - a list of `(index, name)` tuples built with `enumerate`
- `'paired'` - a list of `(name, score)` tuples built with `zip`
- `'ranked'` - the pairs sorted by score from highest to lowest, built with `sorted(..., key=...)` (do not modify `names` or `scores`)
- `'top'` - the name with the highest score; if several names tie, return the first one encountered

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def index_and_pair(names, scores):
    indexed = [(index, name) for index, name in enumerate(names)]
    paired = list(zip(names, scores))
    ranked = sorted(paired, key=lambda pair: pair[1], reverse=True)
    top = max(names, key=lambda name: scores[names.index(name)])
    return {
        "indexed": indexed,
        "paired": paired,
        "ranked": ranked,
        "top": top,
    }
```

**Why:** `enumerate` yields `(index, value)` pairs and `zip` aggregates two iterables into tuples, stopping at the shorter one. `sorted` with a `key` argument leaves the original lists untouched, and `max` returns the first element when there is a tie.

</details>

````python
result = index_and_pair(["ann", "bob", "cid"], [3, 1, 2])
assert result["indexed"] == [(0, "ann"), (1, "bob"), (2, "cid")]
assert result["paired"] == [("ann", 3), ("bob", 1), ("cid", 2)]
assert result["ranked"] == [("ann", 3), ("cid", 2), ("bob", 1)]
assert result["top"] == "ann"
````

## Q14. Coding

Write a context manager class named **`FileManager`**:

- `FileManager(path, mode='w')` stores the file path and the mode.
- `__enter__` opens the file with `open(self.path, self.mode)`, stores the file object on `self.file`, and **returns** the file object so it can be used in a `with` block.
- `__exit__` closes `self.file` if it is not `None` and returns `False`, so that any exception raised inside the `with` block still propagates.

Use it like this:

```
with FileManager("notes.txt", "w") as handle:
    handle.write("hello")
```

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class FileManager:
    def __init__(self, path, mode="w"):
        self.path = path
        self.mode = mode
        self.file = None

    def __enter__(self):
        self.file = open(self.path, self.mode)
        return self.file

    def __exit__(self, exc_type, exc_value, traceback):
        if self.file is not None:
            self.file.close()
        return False
```

**Why:** A context manager needs `__enter__` and `__exit__`. `__enter__` returns the resource that the `with` statement binds to its target, and `__exit__` runs on the way out of the block. Returning `False` (or `None`) from `__exit__` lets exceptions propagate, exactly like the built-in `with open(...)` behaviour.

</details>

````python
import os
import tempfile

folder = tempfile.mkdtemp()
path = os.path.join(folder, "notes.txt")

with FileManager(path, "w") as handle:
    handle.write("hello")

with open(path, "r") as handle:
    content = handle.read()

assert content == "hello"

with FileManager(path, "a") as handle:
    handle.write(" world")
    saved_handle = handle

assert saved_handle.closed is True

try:
    with FileManager(path) as handle:
        handle.write("boom")
        raise ValueError("something went wrong")
except ValueError:
    pass
else:
    raise AssertionError("the exception should have propagated out of the with block")
````
