**Q1 (multiple choice, 1 point)** - What is the type of the result of the expression `3 + 1.5 + 4`?

A. `int`
B. `float`
C. `str`
D. `bool`

<details>
<summary>Answer</summary>

**B. `float`**

**Why:** Python uses dynamic typing, and `int` plus `float` gives a `float`. `type(3 + 1.5 + 4)` returns `<class 'float'>` because one of the operands (`1.5`) is a floating point number.
</details>

**Q2 (multiple choice, 1 point)** - In Python 3, what does `7 // 4` evaluate to?

A. `1.75`
B. `1`
C. `3`
D. `7`

<details>
<summary>Answer</summary>

**B. `1`**

**Why:** `//` is *floor division*: it divides and truncates the decimal without rounding, so `7 // 4` is `1`. For comparison, `7 / 4` is `1.75` (true division, always a float) and `7 % 4` is `3` (the remainder).
</details>

**Q3 (multiple choice, 1 point)** - Which statement about strings is correct?

A. You can change a character in a string with `s[0] = 'x'`
B. Strings are immutable: once a string is created its characters cannot be changed in place, so you build a new string instead
C. Strings can be extended with `.append()`, exactly like lists
D. A string is just a special kind of list

<details>
<summary>Answer</summary>

**B. Strings are immutable: once a string is created its characters cannot be changed in place, so you build a new string instead.**

**Why:** Trying `s[0] = 'x'` raises `TypeError: 'str' object does not support item assignment`. Strings have no `.append()` method (that is a list method), and a string is a sequence, not a list. You can still concatenate with `+` and repeat with `*`, but those create new strings.
</details>

**Q4 (multiple choice, 1 point)** - What does `type((5))` return?

A. `<class 'tuple'>`
B. `<class 'int'>`
C. `<class 'list'>`
D. It raises a `TypeError`

<details>
<summary>Answer</summary>

**B. `<class 'int'>`**

**Why:** Parentheses on their own only group an expression, so `(5)` is just the integer `5`. To build a tuple with a single element you need a trailing comma: `(5,)` has type `<class 'tuple'>`, while `(5)` and `5` are both integers.
</details>

**Q5 (conceptual, 1 point)** - Without running any code, write down the value of each expression for the string `s = 'Hello World'`:

(a) `s[-1]`
(b) `s[:3]`
(c) `s[::2]`
(d) `s[:-1]`

<details>
<summary>Answer</summary>

(a) `'d'`
(b) `'Hel'`
(c) `'HloWrd'`
(d) `'Hello Worl'`

**Why:** A negative index counts from the end of the sequence, so `s[-1]` is the last character. A slice is always "up to, but not including" the upper bound, so `s[:3]` stops before index 3 and gives the first three characters. </details>

**Q7 (conceptual, 1 point)** - A list and a tuple both store ordered items. What is the main difference between them, and what happens when this code runs?

```text
t = (1, 2, 3)
t[0] = 'change'
```

<details>
<summary>Answer</summary>

Lists are **mutable** (their items can be changed, added or removed), while tuples are **immutable**: once a tuple is created it cannot be changed and it cannot grow.

The code raises `TypeError: 'tuple' object does not support item assignment`, because `t[0] = 'change'` tries to modify an item of an immutable tuple. The same code would work fine on a list.

**Why:** Tuples are used when you want a fixed, unchangeable piece of data (for example days of the week or a calendar date), and they offer a little less overhead than lists.
</details>

**Q8 (conceptual, 1 point)** - Consider this file code:

```text
f = open('demo.txt', 'w+')
f.write('Hello')
f.seek(0)
print(f.read())
f.close()
```

What does it print, and why is `seek(0)` needed before `read()`? Also: what does the mode `'w'` do to a file that already exists?

<details>
<summary>Answer</summary>

It prints `Hello`.

`seek(0)` is needed because after `write()` the file cursor sits at the end of the file, so `read()` would return an empty string. `seek(0)` moves the cursor back to the start of the file (index 0) so the content can be read.

Opening with `'w'` (or `'w+'`) **truncates** the file: if `demo.txt` already existed, everything in it is deleted and only the new content remains. Use `'a'` to append to a file, and `'w+'` or `'a+'` if you want to read and write with the same file object (plain `'w'` is write only, so `read()` on it would raise `io.UnsupportedOperation`). Always `close()` the file when you are done with it.

**Why:** A file object keeps a cursor position. `read()` reads from the current position onwards, so after writing you have to move the cursor back with `seek()`.
</details>

**Q10 (coding exercise, 2 points)** - String indexing and slicing.

Remember that strings are immutable, so you cannot change the characters of `s` in place.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
s = 'Complete Python'
first = s[0]
middle = s[1:5]
backwards = s[::-1]
```

**Q13 (coding exercise, 2 points)** - Dictionaries.

Start with `inventory = {'apples': 5, 'oranges': 3}` and then:

1. add the key `'pears'` with the value 2
2. look up the missing key `'bananas'` with `.get('bananas', 0)` and save the result in `missing_default`
3. `update` the dictionary with `{'grapes': 7}`
4. `delete` the key `'oranges'` with `del`

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
inventory = {'apples': 5, 'oranges': 3}
inventory['pears'] = 2

```

</details>

**Q14 (coding exercise, 2 points)** - Sets.

1. Create the list `numbers = [1, 1, 2, 3, 3, 4]` and store its unique values in the set `unique` (use `set(...)`).
2. Create the sets `a = {1, 2, 3, 4, 5}` and `b = {4, 5, 6, 7}`.
3. Build four new sets from `a` and `b`:

- `union`: every element that is in either set (operator `|`)
- `intersection`: the elements the two sets share (operator `&`)
- `difference`: the elements that are in `a` but not in `b` (operator `-`)
- `symmetric_difference`: the elements that are in one set but not in both (operator `^`)

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
numbers = [1, 1, 2, 3, 3, 4]
unique = set(numbers)
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7}
union = a | b
intersection = a & b
difference = a - b
symmetric_difference = a ^ b
```

**Why:** A set keeps only unique elements, so `set(numbers)` drops the repeated 1s and 3s. The set operators give the union, the intersection, the difference and the symmetric difference.

</details>

````python
assert unique == {1, 2, 3, 4} and len(unique) == 4
assert intersection == {4, 5}
assert difference == {1, 2, 3} and symmetric_difference == {1, 2, 3, 6, 7} and union == {1, 2, 3, 4, 5, 6, 7}
````
