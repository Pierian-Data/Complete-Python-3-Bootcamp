### Q6 - Conceptual (1 point)

The course opens CSV files in two different ways: `open('example.csv')` and `open('example.csv', encoding="utf-8")`. What problem does the second form solve, and what error do you hit without it?

Answer in words before you open the solution block below.

<details>
<summary>Solution</summary>

**Correct answer:** without an explicit encoding, Python uses the system's default encoding, and a CSV file that contains characters outside that encoding (an `@` symbol, foreign characters, and so on) raises a `UnicodeDecodeError` with a "can't decode byte ..." message. Passing `encoding="utf-8"` tells Python to read the file as UTF-8, so the file loads correctly.

**Why:** the course shows the failing `list(csv_data)` first and uses the "can't decode" line in the traceback as the giveaway that the problem is an encoding problem, then fixes it with `encoding="utf-8"`.

</details>

### Q7 - Conceptual (1 point)

The course writes CSV files with `open('to_save_file.csv', 'w', newline='')` and then appends to the same file with `open('to_save_file.csv', 'a', newline='')`. Explain what the `'w'` and `'a'` modes do, and why `newline=''` is passed when writing.

Answer in words before you open the solution block below.

<details>
<summary>Solution</summary>

**Correct answer:** `'w'` opens the file for writing and *truncates* it - any existing content is lost, which the course warns about. `'a'` opens the file in append mode and adds new content at the end, keeping what is already there. The `newline=''` argument tells the `csv` module to handle the line endings itself, so you do not get stray blank lines from Python's newline translation.

**Why:** the `csv` module writes the row separator itself, and the docs recommend `newline=''` for exactly this reason - without it, Windows-style newline translation can double up the line endings.

</details>

### Q9 - Coding exercise (2 points)

Read a CSV file with `csv.reader`.

Write the file `people.csv` into the scratch directory (`os.path.join(TMPDIR, 'people.csv')`) with these three rows, using `csv.writer` and `newline=''`:

```text
first_name,last_name,email
Ada,Lovelace,ada@example.com
Alan,Turing,alan@example.com
```

Then read it back with `csv.reader` and store:

- **`rows`** - the list of every row, each row a list of strings
- **`emails`** - a list with just the email address of each data row (skip the header row)

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import csv
import os

path = os.path.join(TMPDIR, 'people.csv')

with open(path, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['first_name', 'last_name', 'email'])
    writer.writerows([['Ada', 'Lovelace', 'ada@example.com'],
                      ['Alan', 'Turing', 'alan@example.com']])
```

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import csv
import os

with open(path, 'w', newline='') as f:

    writer.writerow(['first_name', 'last_name', 'email'])
```

### Q14 - Coding exercise (2 points)

Work with a PDF using `pypdf`.

`pypdf` may not be installed in your environment, so guard your code: wrap the `import pypdf` in a `try`/`except ImportError` block, print `SKIP: pypdf not installed` in the `except` branch and leave `page_count` and `page_text` as `None` there.

If `pypdf` *is* available, then:

1. Create a one page PDF with `pypdf.PdfWriter()`, add a page with `writer.add_blank_page(width=200, height=200)` and write it to `os.path.join(TMPDIR, 'generated.pdf')` in binary write mode.
2. Read it back with `pypdf.PdfReader(...)` and store the number of pages in **`page_count`** and the text of the first page (`reader.pages[0].extract_text()`) in **`page_text`**.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import os

page_count = None
page_text = None

try:
    import pypdf
except ImportError:
    print('SKIP: pypdf not installed')
else:
    path = os.path.join(TMPDIR, 'generated.pdf')
    writer = pypdf.PdfWriter()
    writer.add_blank_page(width=200, height=200)
    with open(path, 'wb') as f:
        writer.write(f)

    reader = pypdf.PdfReader(path)
    page_count = len(reader.pages)
    page_text = reader.pages[0].extract_text()
```

**Why:** `PdfWriter` builds a document and `PdfReader` reads it back; the number of pages is `len(reader.pages)` and the text of a page comes from `extract_text()`. A blank page has no text, so `page_text` is an empty string.

</details>

````python
try:
    import pypdf
except ImportError:
    print('SKIP: pypdf not installed')
else:
    assert page_count == 1
    assert isinstance(page_text, str)
    assert page_text == ''
````
