# Project Portfolio - one project per lesson

A portfolio entry for every course section. Each entry is a small project you
build yourself, in your own repository, that demonstrates the skills of one
section. Done in order, the twenty entries add up to a course-wide portfolio
with the capstone at the centre.

**How this works**

1. Read the entry for your current section.
2. Build the project in its own repository (not in this course repo).
3. Fill in the copy-paste portfolio entry at the bottom of the section and put
   it in your project's README.
4. Tick the deliverables. A project is "portfolio ready" when every box is
   ticked and a stranger can run it in under a minute.

The entries are deliberately small. A finished 200-line project with tests and
a README is worth more than five unfinished ones.

## 1. The portfolio quality bar

Before an entry goes in the portfolio, it must clear these checks:

| Requirement | Why it matters |
| --- | --- |
| A README that says what it is, in the first two lines | Recruiters stop reading after that. |
| A "How to run it" section with the exact commands | Nobody should need to ask you how to start it. |
| A screenshot or a short demo (GIF or terminal recording) | Shows the thing working, not the code. |
| Tests that pass | Proof the code is not a one-off that happened to work. |
| No credentials, keys or personal data in the repo | A hard no in review. |
| Type hints or docstrings on the public functions | Shows you wrote it for other people. |
| A licence file | Makes it usable, and says you thought about it. |
| Meaningful commit history | Evidence of how you actually worked. |

Keep one repository per project. A single repository called "python-exercises"
with twenty folders reads as coursework; twenty small repositories read as a
body of work.

## 2. How your project is graded

A test proves you can use a feature. A project proves you can build with it,
so it is graded differently - on a rubric, not on a pass/fail check.

The same rubric is used for **every** project in the course, which is what
makes the marks comparable. It lives in `rubrics.json` and has six criteria:

| Criterion | Weight | Asks |
| --- | --- | --- |
| Functionality | 3 | Does it do what the brief asks? Are the required features there and working? |
| Robustness | 2 | Does it handle bad input, empty data and edge cases without crashing? |
| Tests | 2 | Is there an automated suite, does it pass from a clean checkout, does it cover the core? |
| Code quality and structure | 2 | Clear names, sensible split, no dead code? |
| Documentation | 2 | Does the README say what it is, how to run it, and show the output? |
| Use of the section's skills | 1 | Does it actually use what this section taught? |

Each criterion is scored 0-4:

| Score | Meaning |
| --- | --- |
| 0 | not attempted |
| 1 | partial - some of it exists, little of it works |
| 2 | basic - the core works, the edges do not |
| 3 | solid - works well, edges handled, tested |
| 4 | excellent - the standard you would show an employer |

The mark is `sum(score x weight) / (4 x total weight)`, and it is recorded in
the same chained ledger as your test scores:

```bash
python 20-Testing-Syllabus/grade_records.py rubric P05
python 20-Testing-Syllabus/grade_records.py record-project P05 --scores 3,4,2,3,4,3
python 20-Testing-Syllabus/grade_records.py record-project P05 --scores 3,4,2,3,4,3 --repo ../my-project
```

The `--repo` flag records the repository path and its current commit, so the
mark is tied to a specific version of your code.

**Get it reviewed before you self-score.** The professor reads your code
against the rubric and writes the review for you:

```bash
python 20-Testing-Syllabus/professor.py review --project P05 --path ../my-project
```

The review says what works, what to change next (in order, naming the files),
the highest-leverage hour of work, and a suggested score per criterion with a
justification. Use it to calibrate your own self-assessment - then record the
score you agree with, and note in `--notes` where the review and your
self-score differed.

## 5. The capstone, in more detail

The capstone (section 18) is the project you put at the top of your portfolio.
Everything else in the index is supporting evidence.

**Specification (the course brief)**

A program that lets you query a movie database - title, release year, rating,
actors and director - and stores what you find in a local file you can reload
and edit.

**Build it in layers**

| Layer | What you add | Done when |
| --- | --- | --- |
| 1. Data model | A `Movie` class and a `MovieDatabase` collection | You can create a movie and read it back |
| 2. Persistence | Load and save to JSON | Closing and reopening the program keeps your data |
| 3. Query | Search by title, year and rating | Every field the brief lists is searchable |
| 4. Interface | A menu or CLI, plus a final summary | A stranger can use it without instructions |
| 5. Tests | Unit tests for model, storage and search | `pytest` passes from a clean checkout |
| 6. Presentation | README, screenshots, demo | The entry is portfolio ready |

**Write the retrospective**

End the README with a short section: what you built, the design decision you
would change, and the bug that took longest. Interviewers ask about the third
one more than the first two.
