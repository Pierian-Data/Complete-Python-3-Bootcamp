# How to Use the Test Suite

Everything you need to run, mark and repeat the tests. Read this once, then
use `S0 - Testing Syllabus.ipynb` as your blueprint.

- `S0 - Testing Syllabus.ipynb` - what is tested, weights, pass bands, study plan.
- `S2 - Grading Records.ipynb` - your live mark sheet: best scores, course mark, band, next target.
- `S3 - Project Portfolio.ipynb` - one portfolio project per course section, with the README entry to copy.
- `S5 - Using the Learning System.ipynb` - the lesson on the whole system: tests, projects, logging, the weekly loop.
- `PRACTICE - Practice Test.ipynb` - a dry run of the test mechanics before you take a real one.
- `PRACTICE - Practice Project - Book Tracker.ipynb` - a practice project with the full brief and rubric.
- `T00` ... `T19` - one marked test per course section.
- `FINAL - Final Comprehensive Exam.ipynb` - the cumulative exam.
- `grade.py` - the self-grader (standard library only, no Jupyter needed).
- `grade_records.py` - logs attempts, computes the course mark, exports `GRADE_REPORT.md`.
- `nbkit.py` - the builder used to generate these notebooks.

## 1. The three kinds of question

| Kind | What it looks like | How it is marked | Points |
| --- | --- | --- | --- |
| Multiple choice | Markdown with options A - D | You, against the collapsed solution | 1 |
| Conceptual / predict the output | Markdown asking for an explanation or the printed output | You, against the collapsed solution | 1 |
| Coding exercise | A code cell for your answer, then a `check` cell of asserts | The grader, automatically | 2 |

The auto-marked half is what makes the suite objective: for every coding
exercise there is an `assert` that is true only if you produced the right
result.

## 2. Anatomy of a coding question

Four cells, always in this order:

1. **Question** (markdown) - the task, and the exact names you must use.
2. **Your answer** (code) - you write here. It starts as a comment.
3. **Solution** (collapsed markdown) - the reference answer plus a "Why:" note.
4. **Check** (code) - the assertions the grader runs.

The solution sits above the check on purpose: the grader walks the notebook
once from top to bottom, so the reference code is already defined by the time
the check runs.

A live example. Run the cell, then the cell below it:

```python
def word_count(text):
    # Return a dict of lowercase word -> count, ignoring punctuation.
    counts = {}
    for raw in text.lower().split():
        word = raw.strip(".,!?;:")
        if word:
            counts[word] = counts.get(word, 0) + 1
    return counts
```

````python
def word_count(text):
    # Return a dict of lowercase word -> count, ignoring punctuation.
    counts = {}
    for raw in text.lower().split():
        word = raw.strip(".,!?;:")
        if word:
            counts[word] = counts.get(word, 0) + 1
    return counts
````

````python
assert word_count("Hello hello world") == {"hello": 2, "world": 1}
assert word_count("") == {}
````

If your implementation is wrong, the check fails and tells you so:

```
AssertionError: assert {'hello': 1, 'world': 1} == {'hello': 2, 'world': 1}
```

That message is the difference between "your code is wrong" and "your code is
wrong in this exact way". Read it before rewriting anything.

## 3. Running a test in JupyterLab

Before your first real test, run `PRACTICE - Practice Test.ipynb` - it is a dry run
of exactly these mechanics, and it does not count towards the mark.

1. Open the notebook from the file browser in JupyterLab.
2. Pick the `Python 3 (ipykernel)` kernel (Anaconda's is fine).
3. Work top to bottom. Write your answer in the code cell, then run it with
   `Shift+Enter`.
4. Run the check cell. A green tick means that exercise is right.
5. Run `Cell > Run All` from the menu once you have finished a test, so the
   grader sees your answers in a consistent order.

**Restart the kernel** (`Kernel > Restart Kernel`) between attempts if a cell
has produced a confusing state. Note that restarting also clears your answers,
so save the notebook (`Ctrl+S`) first.

`Runtime > Restart Kernel and Run All Cells` is the fastest way to see the
whole test at once, but it will fail every check you have not answered yet.
That is expected, not a bug.

## 4. Grading with `grade.py`

The grader replays a notebook's code cells in order in one namespace, exactly
like a kernel, and reports every cell tagged `check`.

From the repository root:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade.py "T03 - Methods and Functions Test.ipynb"
```

```
[OK ] T03 - Methods and Functions Test.ipynb                       6/6 checks passed

0 checks passed, 0 failed across 1 notebooks.
```

- `PASS` - the assertions held.
- `FAIL` - the assertion did not hold; the message is the exception raised.
- The script exits with status 1 if any check failed, so you can wire it into a
  script or CI job.

Grade every test in the folder (no arguments):

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade.py
```

Validate the test suite itself, by running the reference solutions and
confirming every check passes:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade.py --solutions
```

Use `--solutions` if you want to know whether a failing check is your mistake
or a broken test. It should report 0 failures for every notebook.

## 5. Reading the output

| Output | Meaning | What to do |
| --- | --- | --- |
| `[OK ] name 6/6 checks passed` | Every check passed | Mark the coding half as correct, then self-mark the prose questions. |
| `[BAD] name 4/6 checks passed` plus `FAIL Qn` | Some checks failed | The failing line names the exercise; open that question and fix it. |
| `FAIL Qn: NameError: name 'x' is not defined` | The identifier was never created | You did not write the answer, or you used a different name than the question asked for. |
| `FAIL Qn: AssertionError` | The answer exists but the value is wrong | Compare with the solution in the collapsed block above the check. |
| `PROBLEM cell n failed: ...` | A solution block itself raised | A broken test - re-run with `--solutions` to confirm. |
| `SKIP: pypdf not installed` | Optional dependency missing | Expected on sections 13 - 16 without extra packages. Not a failure. |

**Naming matters.** The question always states the identifier to use
("define `add(a, b)`", "assign the result to `total`"). The check asserts on
that exact name, so `Add` or `result` will not satisfy it.

## 6. Scoring your own work

The grader covers the coding half. For the prose half, use the solution
blocks: read your answer, read the solution, and mark them as:

- **2/2** - you got it and can explain why.
- **1/2** - right idea, missing detail.
- **0/2** - wrong.

Keep a running total somewhere convenient - a copy of the readiness checklist
in `S0 - Testing Syllabus.ipynb` works well. A section is signed off at 14 / 20.

## 7. Troubleshooting

**`python` is not recognised.** Use the full path to Anaconda's interpreter:
`C:\Users\andre\anaconda3\python.exe`. Or open the Anaconda Prompt and run
`grade.py` from there.

**The grader cannot find the notebook.** Paths are resolved relative to the
`20-Testing-Syllabus` folder, so the file name alone is enough:
`grade.py "T05 - Object Oriented Programming Test.ipynb"`.

**A check fails even though your code looks right.** Two usual causes: the
answer cell was never run (the grader replays from scratch, so it runs the
cells as they are saved - save with `Ctrl+S` first), or you used a different
identifier than the question specified.

**Your answers disappeared.** The kernel was restarted. Save before you
restart, and keep scratch work in a separate notebook.

**`ModuleNotFoundError` for `requests`, `bs4`, `PIL`, `pypdf` or `openpyxl`.**
Sections 13 - 16 use packages from the course. The checks for those sections
are written to skip themselves when the package is absent, so the test still
marks cleanly. Install them with
`pip install requests beautifulsoup4 pillow pypdf openpyxl` in the Anaconda
prompt if you want the full marks for those sections.

**A test hangs.** Something in your answer is waiting for input. Interrupt the
cell (`Kernel > Interrupt Kernel`) and look for `input()` in your code.

## 8. Logging the attempt and your mark sheet

The grader marks the coding half; the ledger keeps the record. After you finish
a test, log the attempt:

```

```

That grades the notebook as it is saved, counts the checks that pass, asks for

choice or conceptual question) and writes the attempt to `records.json`.

Open `S2 - Grading Records.ipynb` to see the same data rendered: best score
per module, the weighted course mark, the band, and the next module to attack.


which writes `GRADE_REPORT.md` next to the notebooks. `S3 - Project Portfolio.ipynb`
describes the project you build alongside each module.

## 9. Ground rules for the suite

- Every check is deterministic: same result every run. No network, no
  randomness without a fixed seed, no dependence on today's date.
- Nothing in the suite writes into the repository. File-based exercises use
  `tempfile` and clean up after themselves.
- No test needs credentials, API keys or an email account.
- Retakes are unlimited and free. The final attempt is the one that counts.

## 10. Adding your own test notebooks

The notebooks here are generated, so they can be extended without hand-editing
JSON. `nbkit.py` writes a valid notebook from a list of cells:

```python
import sys
sys.path.insert(0, r"C:/Users/andre/Learning/Udemy-Complete-Python-3-Bootcamp/20-Testing-Syllabus")
from nbkit import md, code, check, solution, nb

cells = [
    md("### Q1. Add two numbers\n\nDefine `add(a, b)` returning the sum."),
    code("# your answer here"),
    solution("def add(a, b):\n    return a + b", "Arguments are positional by default."),
    check("assert add(2, 3) == 5", qid="Q1"),
]

nb(r"C:/Users/andre/Learning/Udemy-Complete-Python-3-Bootcamp/20-Testing-Syllabus/T99 - My Extra Test.ipynb", cells)
```

Build and validate:

```
C:\Users\andre\anaconda3\python.exe C:/Users/andre/AppData/Local/Temp/nbbuild/my_test.py
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus/grade.py --solutions "T99 - My Extra Test.ipynb"
```

`--solutions` must report 0 failures, otherwise the check and the reference
solution disagree and the test is not fit to hand out.
