### Q1 (multiple choice, 1 point)

What is the value of `2 ** 3 ** 2` in Python?

A. 64
B. 512
C. 12
D. 8

**Correct answer: B. 512**

**Why:** Exponentiation is right-associative, so the expression is grouped as
`2 ** (3 ** 2)`, that is `2 ** 9`, which is 512. If you wanted `(2 ** 3) ** 2`
you would have to write the parentheses yourself, giving 64.

### Q2 (predict the output, 1 point)

What exactly does this code print?

```text
s = 'hello'
print(s[1:4])
print(s[::-1])
```

**Correct answer:**

```text
ell
olleh
```

**Why:** `s[1:4]` takes the characters at index 1, 2 and 3 (the end index is
excluded), giving `ell`. `s[::-1]` steps backwards through the whole string with a
step of -1, which reverses it, giving `olleh`.

### Q3 (multiple choice, 1 point)

Which format string prints the number `3.14159` with two decimal places, right
aligned in a field of 8 characters?

A. `'{0:.2f}'`
B. `'{0:8.2f}'`
C. `'{0:2f}'`
D. `'{0:.8f}'`

**Correct answer: B. `'{0:8.2f}'`**

**Why:** The part after the colon is `8.2f`: `8` is the minimum field width and
`.2` is the number of decimals. The `f` conversion keeps two decimal places and
right aligns the result inside the field. `.2f` alone (A) prints without the
padding, and `2f` (C) sets the width to 2, which is too narrow for the value.

### Q31 (coding exercise, 5 points)

Write a function called `build_inventory(lines)`.

`lines` is a list of strings. Each meaningful line looks like `'apple: 3'`, but
the data is messy: there may be extra spaces, and some lines are blank or do
not contain a colon at all.

Rules:

1. Skip any line that is blank (only whitespace) or that contains no `:`.
2. Split each remaining line on the **first** colon only.
3. Strip whitespace from the item name and from the count, and convert the
   count with `int()`.
4. Build a dictionary mapping the item name to the total count. If the same
   item appears more than once, add the counts together.
5. Return a tuple `(inventory, names)` where `inventory` is that dictionary and
   `names` is the list of the item names sorted alphabetically.

Example: `build_inventory(['apple: 3', '  banana :2', 'apple: 1'])` should give
`({'apple': 4, 'banana': 2}, ['apple', 'banana'])`.

Write your solution in the next cell.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def build_inventory(lines):
    inventory = {}

    for line in lines:
        if ':' not in line:
            continue

        name, _, count = line.partition(':')
        name = name.strip()
        count = count.strip()

        if not name or not count:
            continue

        value = int(count)
        inventory[name] = inventory.get(name, 0) + value

    names = sorted(inventory.keys())
    return inventory, names
```

**Why:** A line that has no colon (or is blank once stripped) is
skipped with `continue`. `str.partition` splits on the first colon only and
returns the text before it, the separator and the text after it, which is
exactly what is needed when the value itself could contain a colon.
`inventory.get(name, 0)` adds the counts of repeated items together, and
`sorted(inventory.keys())` gives the alphabetical list of names.

</details>

Example: `make_order('apple', 'pear', quantity=2, currency='EUR', discount=0.25)`

```python
def make_order(*items, quantity=1, currency='$', **options):

    count = len(items)
```

Create a class called `Rectangle` with the following behaviour:

2. `__str__(self)` returns the string `'Rectangle(w=3, h=4)'` for a rectangle

   `f"Rectangle(w={self.width}, h={self.height})"`.
3. `__len__(self)` returns the area of the rectangle as an `int`, so that
   `len(Rectangle(3, 4))` is `12`.
4. `__eq__(self, other)` returns `True` when `other` is a `Rectangle` with the
   same area, and `False` in every other case (including when `other` is not a
   rectangle at all). It must never raise an exception.

Write your solution in the next cell.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle(w={self.width}, h={self.height})"

    def __len__(self):
        return int(self.width * self.height)

    def __eq__(self, other):
```

### Q34 (coding exercise, 5 points)

Write a decorator called `count_calls` that wraps a function and counts how
many times it has been called.

Rules:

1. `count_calls(func)` returns a wrapper function.
2. The wrapper must forward every positional and keyword argument to the
   original function and return whatever the original function returns.
3. The wrapper must expose an integer attribute called `calls`, starting at
   `0`, which is incremented on every call. So after `add(2)` the attribute
   `add.calls` is `1`.
4. The wrapper must keep the original function's `__name__` and `__doc__`
   (use `functools.wraps`).

Use it like this:

```text
@count_calls
def add(a, b=1):
    return a + b
```

Write your solution in the next cell.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import functools


def count_calls(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)

    wrapper.calls = 0
    return wrapper
```

**Why:** The decorator builds a closure over `func`. The
wrapper takes `*args` and `**kwargs` so it can pass any call on unchanged, and
`wrapper.calls` is created once on the wrapper object and incremented on every
call. `functools.wraps` copies the name, docstring and other metadata of the
original function onto the wrapper.

</details>

````python
@count_calls
def add(a, b=1):
    """Add two numbers."""
    return a + b

assert add(2) == 3
assert add.calls == 1
assert add(2, b=5) == 7
assert add.__name__ == 'add' and add.calls == 2
````

### Q35 (coding exercise, 5 points)

Write a function called `safe_divide(a, b, log=None)` that divides safely.

Rules:

1. If `b` is not zero, return the result of `a / b` (a float).
2. If `b` is zero, catch the `ZeroDivisionError` and return the string
   `'undefined'`.
3. In a `finally` block, append the string `'done'` to the list `log`, so the
   log always records that an attempt was made. If `log` is `None`, create a
   new empty list and use that.
4. Only `ZeroDivisionError` may be caught: a `TypeError` (for example from
   dividing a string by a number) must still propagate to the caller.

Example: `safe_divide(6, 3)` returns `2.0` and `safe_divide(1, 0)` returns
`'undefined'`.

Write your solution in the next cell.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def safe_divide(a, b, log=None):
    if log is None:
        log = []

    try:
        return a / b
    except ZeroDivisionError:
        return 'undefined'
    finally:
        log.append('done')
```

**Why:** The `finally` block runs whether the division
succeeded or the exception was handled, so the log always gets its entry.
Only `ZeroDivisionError` is caught, so a `TypeError` from a non-numeric
argument still travels up to the caller.

</details>

````python
log = []
assert safe_divide(6, 3) == 2.0
assert safe_divide(1, 0, log) == 'undefined'
assert log == ['done']

raised = False
try:
    safe_divide('a', 1)
except TypeError:
    raised = True
assert raised is True
````

### Q36 (coding exercise, 5 points)

Write a function called `top_scores(students, n)`.

`students` is a list of `(name, score)` tuples.

Rules:

1. Use a list comprehension to build the list of the `n` highest scores.
2. Use `sorted(..., key=...)` to order the result: highest score first, and
   when two scores are equal the names are ordered alphabetically.
3. Return the list of the chosen names.
4. Return an empty list when `n` is zero or negative, or when `students` is
   empty.

Example: with `students = [('ana', 88), ('bob', 95), ('cleo', 88), ('dan', 95)]`,
`top_scores(students, 3)` gives `['bob', 'dan', 'ana']`.

Write your solution in the next cell.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def top_scores(students, n):
    if n <= 0:
        return []

    ordered = sorted(students, key=lambda student: (-student[1], student[0]))
    best = ordered[:n]

    return [name for name, score in best]
```

**Why:** Sorting by `(-score, name)` puts the
highest score first and, for equal scores, the alphabetically first name
first. The list comprehension then pulls the names out of the top `n` tuples.

</details>

````python
students = [('ana', 88), ('bob', 95), ('cleo', 88), ('dan', 95)]
assert top_scores(students, 2) == ['bob', 'dan']
assert top_scores(students, 3) == ['bob', 'dan', 'ana']
assert top_scores(students, 0) == []
assert top_scores([], 3) == []
````
