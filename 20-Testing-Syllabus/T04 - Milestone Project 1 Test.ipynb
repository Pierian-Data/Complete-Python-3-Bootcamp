**Q1 (coding exercise, 2 points)** - Warm-up: reversing a string.

Write a function `reverse_string(text)` that returns the string `text` written backwards, so `reverse_string('hello')` returns `'olleh'`.

You may use slicing (`text[::-1]`), which you met in the strings and lambda lectures.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def reverse_string(text):
    return text[::-1]
```

**Why:** A slice with a negative step walks the string backwards, so `text[::-1]` is the reversed string. The course's lambda lecture uses exactly this trick: `lambda s: s[::-1]`.

</details>

````python
assert reverse_string('hello') == 'olleh'
assert reverse_string('') == ''
assert reverse_string('ab') == 'ba'
````

**Q2 (coding exercise, 2 points)** - Warm-up: turning money into a number.

Write a function `money_to_number(amount)` that takes a **string** such as `'$1,234.56'` and returns it as a `float`, so `money_to_number('$1,234.56')` returns `1234.56`.

Remove the leading dollar sign and the thousands separators, then convert the rest with `float()`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def money_to_number(amount):
    cleaned = amount.replace('$', '').replace(',', '')
    return float(cleaned)
```

**Why:** `str.replace()` returns a new string with the dollar sign and the commas removed, and `float()` then converts the remaining digits. Chaining the two `replace()` calls keeps it to one expression.

</details>

````python
assert money_to_number('$1,234.56') == 1234.56
assert money_to_number('$10') == 10.0
assert money_to_number('0.99') == 0.99
````

**Q3 (coding exercise, 2 points)** - Warm-up: flipping a coin.

Write a function `flip_coin()` that returns the string `'Heads'` or `'Tails'`, chosen at random.

Use the `random` module you met in `05-Useful-Operators.ipynb`: `random.randint(0, 1)` returns `1` or `0`. Call `random.seed(0)` once, at the top of the cell, **before** you define the function, so the result is reproducible.

The check cell resets the seed itself, so it always sees the same sequence.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import random

random.seed(0)

def flip_coin():
    if random.randint(0, 1) == 1:
        return 'Heads'
    return 'Tails'
```

**Why:** `random.randint(0, 1)` gives an inclusive random integer in `[0, 1]`. Seeding the module first makes the sequence reproducible, which is why the check cell can compare against a known result.

</details>

````python
import random

random.seed(0)
first = flip_coin()
random.seed(0)
second = flip_coin()

assert first == second
assert first in ('Heads', 'Tails')
assert flip_coin() in ('Heads', 'Tails')
````

**Q4 (coding exercise, 2 points)** - Warm-up: a simple calculator.

Write a function `calculate(num1, operator, num2)` that performs one arithmetic operation and returns the **number** result:

- `'add'` -> `num1 + num2`
- `'subtract'` -> `num1 - num2`
- `'multiply'` -> `num1 * num2`
- `'divide'` -> `num1 / num2`

For any other operator string, return the string `'unknown operator'`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def calculate(num1, operator, num2):
    if operator == 'add':
        return num1 + num2
    elif operator == 'subtract':
        return num1 - num2
    elif operator == 'multiply':
        return num1 * num2
    elif operator == 'divide':
        return num1 / num2
    else:
        return 'unknown operator'
```

**Why:** An `if` / `elif` chain tests the operator string one value at a time, and the `else` branch is the fallback for anything unexpected.

</details>

````python
assert calculate(6, 'add', 3) == 9
assert calculate(6, 'subtract', 3) == 3
assert calculate(6, 'multiply', 3) == 18
assert calculate(6, 'divide', 3) == 2
assert calculate(6, 'power', 3) == 'unknown operator'
````
