## Q1. Multiple choice

In Python 3, what does `map(function, iterable)` return?

A. A list containing one result per element
B. An iterator that yields one result at a time
C. A tuple of results
D. The original list, modified in place

<details>
<summary>Solution</summary>

**Answer: B** - an iterator that yields one result at a time.

`map()` is lazy: it returns a `map` object (an iterator) and only computes a result when that
result is requested. To see the values right away you wrap it in `list(...)`, exactly as the
course notebooks do.

**Why:** the same rule applies to `filter()` and `zip()` - all three return iterators, which is
why the course writes `list(map(...))`, `list(filter(...))` and `list(zip(...))` in every example.

</details>

## Q2. Multiple choice

Which statement about `filter(function, iterable)` is correct?

A. It keeps the elements for which `function` returns a truthy value
B. It keeps the elements for which `function` returns `False`
C. It removes duplicate elements and returns a list
D. It always returns a list, never an iterator

<details>
<summary>Solution</summary>

**Answer: A** - it keeps the elements for which `function` returns a truthy value.

The function is applied to every element and only the truthy results survive. `filter()` returns
an iterator, so the course always casts the result with `list(...)`.

**Why:** this is the direct counterpart of `map()`: `map` transforms every element, `filter` keeps
a subset of the elements.

</details>

## Q3. Multiple choice

What happens when you zip two iterables of different lengths, for example
`list(zip([1, 2, 3], [4, 5, 6, 7, 8]))`?

A. The result has five pairs and the shorter list is padded with `None`
B. The result has three pairs, because `zip` stops at the shortest iterable
C. `zip` raises a `ValueError`
D. The result has three pairs and the longer list is padded with `None`

<details>
<summary>Solution</summary>

**Answer: B** - three pairs, because `zip` stops at the shortest iterable.

`zip([1, 2, 3], [4, 5, 6, 7, 8])` gives `[(1, 4), (2, 5), (3, 6)]`.

**Why:** as the course notes, only use `zip` with unequal length inputs when you do not care
about the trailing unmatched values.

</details>

## Q4. Multiple choice

What does `list(enumerate(['a', 'b'], start=3))` return?

A. `[(0, 'a'), (1, 'b')]`
B. `[(1, 'a'), (2, 'b')]`
C. `[(3, 'a'), (4, 'b')]`
D. `[(3, 'a'), (3, 'b')]`

<details>
<summary>Solution</summary>

**Answer: C** - `[(3, 'a'), (4, 'b')]`.

`enumerate()` returns `(count, element)` tuples. The counter starts at `0` by default and the
optional `start` argument overrides that default, here counting from 3.

**Why:** the counter is a plain integer that increases by one per element, so it always equals
`start + index`.

</details>

## Q5. Conceptual - predict the output

What does this code print?

```text
from functools import reduce
result = reduce(lambda x, y: x + y, [1, 2, 3, 4])
print(result)
```

<details>
<summary>Solution</summary>

**Output:**

```text
10
```

`reduce(function, sequence)` applies the function to the first two elements, then applies it to
the previous result and the next element, and keeps going until a single value is left.

**Why:** `reduce` collapses a sequence into one value, so its result is a number, not a list.
Note that `reduce` lives in the `functools` module, not in the built-ins.

</details>

## Q6. Conceptual - predict the output

What does this code print?

```text
lst = [True, True, False]
print(all(lst))
print(any(lst))
print(all(x > 0 for x in [1, 2]))
print(any(x > 5 for x in [1, 2]))
```

<details>
<summary>Solution</summary>

**Output:**

```text
False
True
True
False
```

`all()` is `True` only when every element is truthy, `any()` is `True` when at least one element is
truthy. A generator expression works as the iterable and is evaluated lazily.

**Why:** both functions short-circuit - `all()` stops at the first falsy element and `any()` stops
at the first truthy element, so they never scan the whole iterable once the answer is known.

</details>

## Q9. Coding - word lengths with `map`

Use `map()` to create a function `word_lengths(phrase)` that finds the length of each word in the
phrase (words are separated by spaces) and returns the lengths as a list of integers.

The function takes a string and returns a list of integers.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def word_lengths(phrase):
    return list(map(len, phrase.split()))
```

**Why:** `map(len, phrase.split())` applies the built-in `len` to every word, and `list(...)` collects the results.

</details>

````python
assert word_lengths('How long are the words in this phrase') == [3, 4, 3, 3, 5, 2, 4, 6]
assert word_lengths('') == []
assert word_lengths('a bb ccc') == [1, 2, 3]
````

## Q10. Coding - digits to a number with `reduce`

Use `reduce()` to write a function `digits_to_num(digits)` that takes a list of digits and returns
the number they represent. For example `[1, 2, 3]` corresponds to one hundred twenty three.

Do not convert the integers to strings.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
from functools import reduce

def digits_to_num(digits):
    return reduce(lambda x, y: x * 10 + y, digits)
```

**Why:** Each step shifts the accumulator left by one decimal place (`x * 10`) and adds the next digit.

</details>

````python
assert digits_to_num([3, 4, 3, 2, 1]) == 34321
assert digits_to_num([1, 2, 3]) == 123
assert digits_to_num([7]) == 7
````

## Q11. Coding - filter words by first letter

Use `filter()` to write a function `filter_words(word_list, letter)` that returns the words from
`word_list` which start with the target letter `letter`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def filter_words(word_list, letter):
    return list(filter(lambda word: word[0] == letter, word_list))
```

**Why:** `filter` keeps every element for which the lambda returns `True`; `list(...)` turns the iterator into a list.

</details>

````python
words = ['hello', 'are', 'cat', 'dog', 'ham', 'hi', 'go', 'to', 'heart']
assert filter_words(words, 'h') == ['hello', 'ham', 'hi', 'heart']
assert filter_words(words, 'z') == []
assert filter_words(words, 'a') == ['are']
````

## Q12. Coding - concatenate two lists with `zip`

Use `zip()` and a list comprehension to write a function `concatenate(L1, L2, connector)` that
returns a list of the same length as the shorter list, where each value is the string from `L1`
and the matching string from `L2` joined by `connector`.

For example `concatenate(['A', 'B'], ['a', 'b'], '-')` should give `['A-a', 'B-b']`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def concatenate(L1, L2, connector):
    return [word1 + connector + word2 for (word1, word2) in zip(L1, L2)]
```

**Why:** `zip` pairs the elements of both lists and the comprehension joins each pair with the connector.

</details>

````python
assert concatenate(['A', 'B'], ['a', 'b'], '-') == ['A-a', 'B-b']
assert concatenate(['x'], ['y', 'z'], '|') == ['x|y']
assert concatenate([], ['a'], '-') == []
````

## Q13. Coding - index dictionary with `enumerate`

Use `enumerate()` to write a function `d_list(L)` that returns a dictionary whose keys are the
values of the list `L` and whose values are the index of each item.

You may assume that a value appears only once in the given list.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def d_list(L):
    return {key: value for value, key in enumerate(L)}
```

**Why:** `enumerate` yields `(index, element)` pairs, so the comprehension swaps them to build `{element: index}`.

</details>

````python
assert d_list(['a', 'b', 'c']) == {'a': 0, 'b': 1, 'c': 2}
assert d_list(['x']) == {'x': 0}
assert d_list([]) == {}
````

## Q14. Coding - count items equal to their index

Use `enumerate()` to write a function `count_match_index(L)` that returns the count of the number
of items in the list whose value equals its index.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def count_match_index(L):
    return len([num for count, num in enumerate(L) if num == count])
```

**Why:** The comprehension keeps the items whose value equals the index reported by `enumerate`, and `len` counts them.

</details>

````python
assert count_match_index([0, 2, 2, 1, 5, 5, 6, 10]) == 4
assert count_match_index([]) == 0
assert count_match_index([0, 1, 2]) == 3
````
