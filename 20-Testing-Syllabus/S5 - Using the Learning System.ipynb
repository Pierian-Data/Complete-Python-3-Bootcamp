## 1. What this system is

Five things, wired together.

- **A course.** Twenty numbered folders at the root of the repository, one per section, each a set of Jupyter lecture notebooks. `00-Python Object and Data Structure Basics/` through `19-Bonus Material - Introduction to GUIs/`.
- **A test suite.** One marked test per section, `T00 - Objects and Data Structures Test.ipynb` through `T19 - GUI Bonus Material Test.ipynb`, plus the cumulative `FINAL - Final Comprehensive Exam.ipynb`. Each test is 20 points: 4 multiple choice, 4 conceptual, 6 coding exercises.
- **A mark sheet.** `grade_records.py` writes every attempt into `records.json`, a hash-chained ledger, and computes your weighted course mark from it. `S2 - Grading Records.ipynb` renders the same data.
- **A portfolio.** `S3 - Project Portfolio.ipynb` gives you one small build project per section, each with deliverables and a README entry to copy. The projects are the evidence; the marks are the receipt.
- **A professor.** `PROFESSOR.md` at the repository root is the operating manual for an AI tutor, `AGENTS.md` makes any coding agent in this repository behave like that tutor, and `20-Testing-Syllabus/professor.py` is the same tutor as a command line tool.

The point of putting them together is that each piece checks the others. The grader checks your code against assertions nobody can quietly weaken, because `integrity.py manifest` hashes every file in the suite. The ledger checks that your history was not rewritten, because every entry carries the hash of the one before it and `SEAL.json` signs the head of the chain. The portfolio checks that the work exists, because a project with no README and no tests is not a project.

## 3. How to take a test

The mechanics, in order. `S1 - How to Use the Test Suite.ipynb` covers the same ground in more detail; this is the version you follow end to end.

**Step 1. Open the test for your section.** `T00 - Objects and Data Structures Test.ipynb` for section 00, `T05 - Object Oriented Programming Test.ipynb` for section 05. Read the header first: it lists the lecture notebooks the test covers, the number of questions, the points and the time budget.

**Step 2. Read the question, then write the answer.** Every coding question is four cells, always in this order:

1. **Question** (markdown): the task, and the exact identifier names you must create.
2. **Your answer** (code): starts as `# your answer here`.
3. **Solution** (collapsed markdown): the reference answer and a "Why:" note.
4. **Check** (code, tagged `check`): the assertions the grader runs.

**Step 3. Run your answer cell** with `Shift+Enter`. Read what it printed before you move on.

**Step 4. Run the check cell.** No output and no red traceback means the exercise is right. A red traceback means the assertions did not hold, and the message tells you which one.

**Step 5. Mark the prose questions.** The multiple choice and conceptual questions are yours to mark: read your answer, open the collapsed solution next to the question, and award 1 point for a correct answer with a reason you could give out loud, 0 for the rest.

**Step 6. Log the attempt** so it counts towards the mark sheet. From the repository root:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade_records.py run "T00 - Objects and Data Structures Test.ipynb" --prose 6
```

`run` grades the notebook as it is saved, counts the checks that pass, adds the `--prose` points for the questions you marked by hand, and appends the attempt to the ledger. Leave `--prose` off and it asks you for the number interactively.

Before step 6, grade the notebook on its own and read the failures:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade.py "T00 - Objects and Data Structures Test.ipynb"
```

```
[BAD] T00 - Objects and Data Structures Test.ipynb                4/6 checks passed
        FAIL Q9: AssertionError
        FAIL Q12: NameError: name 'label' is not defined

6 checks passed, 2 failed across 1 notebooks.
```

Each `FAIL` line names the question and the exception the check raised: `AssertionError` means the value was wrong, `NameError` means the identifier never existed. `PROBLEM cell n failed: ...` means a cell itself broke - a broken test rather than a wrong answer, which you can confirm with `--solutions`.

The script exits with status 1 if any check failed, so you can also run it from a script or a CI job.

**What the check cells actually assert.** A check cell is a list of `assert` statements against the names the question asked for. Here is the whole idea in miniature:

```python
def initials(full_name):
    # Return the initials of a full name, e.g. "Ada Lovelace" -> "AL"
    parts = full_name.split()
    return "".join(part[0].upper() for part in parts if part)
```

The matching check cell would be:

```text
assert initials("Ada Lovelace") == "AL"
assert initials("Grace Brewster Murray Hopper") == "GBMH"
```

**Why the identifier names matter.** The check asserts on the exact name in the question. If the question says "create `square`", the check says `assert square == 49`; name it `sq` and the check fails with `NameError: name 'square' is not defined` even though your arithmetic was right. The grader replays the notebook from the top in one namespace, exactly like a kernel, so a name created in an earlier cell is still there when a later check runs, and a name you never created is not.

**When a check fails, read the message before you rewrite anything.** The message is the difference between "your code is wrong" and "your code is wrong in this exact way":

````python
# The two ways a check cell fails: a wrong value, and a missing name.
import traceback


def initials(full_name):
    # Return the initials of a full name, e.g. "Ada Lovelace" -> "AL"
    parts = full_name.split()
    return "".join(part[0].upper() for part in parts if part)


assert initials("Ada Lovelace") == "AL"
assert initials("Grace Brewster Murray Hopper") == "GBMH"
print("both checks passed")

# Now the failure message a check cell would give you.
try:
    assert initials("Ada") == "A.L."
except AssertionError:
    print("FAIL Q9: %s" % traceback.format_exc().strip().splitlines()[-1])

# And the failure message for the wrong identifier.
try:
    assert sq == "AL"
except NameError:
    print("FAIL Q9: %s" % traceback.format_exc().strip().splitlines()[-1])
````

The first failure says your logic is wrong. The second says your code never ran under the name the question asked for. Fixing the second by renaming the variable is the single most common repair in this suite.

**If a check passes in Jupyter but fails in the grader**, the answer cell was probably never saved. The grader reads the file on disk, not the live kernel. Press `Ctrl+S` in JupyterLab, or `File > Save Notebook`, then run the grader again.

**If you suspect the test itself is broken** rather than your answer, run it against the reference solutions. That mode executes the collapsed solution blocks before the check cells:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade.py --solutions "T00 - Objects and Data Structures Test.ipynb"
```

Every check must pass there. If one does not, the check and the reference solution disagree, and that is a bug in the test worth reporting.

Read it line by line:

`grade_records.py report` writes the same information to `GRADE_REPORT.md`, which is the file to commit into your portfolio repository as proof of work. `S2 - Grading Records.ipynb` renders the same ledger as live notebook output, including the sign-off checklist and the attempt history.

````python
# The ledger, read live. This cell only reads; it never writes.
import os
import sys


def find_suite(start):
    # Walk up from `start` looking for the 20-Testing-Syllabus folder.
    current = os.path.abspath(start)
    while True:
        if os.path.exists(os.path.join(current, "grade_records.py")):
            return current
        candidate = os.path.join(current, "20-Testing-Syllabus")
        if os.path.exists(os.path.join(candidate, "grade_records.py")):
            return candidate
        parent = os.path.dirname(current)
        if parent == current:
            return None
        current = parent


SUITE = find_suite(os.getcwd())
print("suite folder:", SUITE or "not found - start Jupyter from the repository root")
if SUITE:
    if SUITE not in sys.path:
        sys.path.insert(0, SUITE)
    import grade_records as gr

    summary = gr.summarize()
    print("Course mark : %.1f / 100 (%s)" % (summary["course_mark"] * 100, summary["band"]))
    print("Signed off  : %d / 20 modules" % len(summary["signed_off"]))
    print("Next target : %s" % (summary["next_target"] or "none - take the final exam"))
    print("Attempts    : %d" % len(summary["attempts"]))
    print("Projects    : %d" % len(summary["projects"]))
    print("Journal     : %d" % len(summary["journal"]))
````

## 5. How to redo a test

Retakes are free, unlimited and never punished. You are not asked to hide a bad attempt; the ledger is built to show the improvement.

- **Nothing is deleted.** `records.json` is append-only.
- **A redo is a new entry that names the one it supersedes.** The new entry carries an attempt number and a `supersedes` reference to the old entry's sequence number.
- **The best attempt counts.** A worse retake can never lower your mark; `grade_records.py` takes the highest points per module.
- **`reset` voids, it does not erase.** `grade_records.py reset T03` appends a `reset` event. The earlier attempts stay in the chain, marked void, and stop counting. Use it when a test was broken, not to hide a bad score.

  Two things to know about `reset`. First, `grade_records.py list` shows the voiding explicitly, so the history reads honestly:

  ```
  2026-11-06  T03       VOIDED by reset - the history above is kept
  ```

  Second, the void applies to the test id, not to a date: once T03 has been reset, later attempts at T03 are voided as well, because the ledger records that T03 was reset rather than which attempt was bad. If you want a bad mark off the record but the module still countable, retake it instead of resetting it.

So the loop is: fix the failing checks, run the test again, log it again. Nothing to clean up.

`grade_records.py list` shows every entry, and the shape of one line tells you the whole story:

## 6. How to do a project

A test proves you can *use* a feature. A project proves you can *build with* it. The test asks "does `range(0, 6)` give five items?"; the project asks "here is a domain with rules, write the code that holds them". The test is marked by assertions; the project is marked against a rubric, because "does it work" and "is it good" are different questions.

The rubric is in `rubrics.json`, and it is the same six criteria for every project, which is what makes project marks comparable across the course:

| Criterion | Weight | The question it asks |
| --- | --- | --- |
| Functionality | 3 | Does it do what the brief asks for? Are the required features present and working? |
| Robustness | 2 | Does it handle bad input, empty data, missing files and edge cases without crashing? |
| Tests | 2 | Is there an automated test suite, does it pass from a clean checkout, and does it cover the core behaviour? |
| Code quality and structure | 2 | Are names, functions and modules clear? Is the code split sensibly? Any dead code or duplicated logic? |
| Documentation | 2 | Does the README say what it is, how to run it, and show the output? Are the public functions documented? |
| Use of the section's target skills | 1 | Does the project actually use the techniques this section taught? |

Each criterion is scored 0 to 4: 0 not attempted, 1 partial, 2 basic, 3 solid, 4 excellent. The weights make the mark, so functionality counts for three times what the section skills count for.

**The five stages.**

1. **Read the brief.** Open the project's rubric first, then the brief:

   ```
   C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade_records.py rubric P05
   ```

   ```
   RUBRIC - Domain Model Library (P05)
   ======================================================================
   A domain model of a real system (employees, products, orders) expressed with classes, dunder methods, an inheritance hierarchy and polymorphism.

   1. Functionality  (weight 3, scored 0-4)
      Does the project do what the brief asks for? Are the required features all present and working?
   2. Robustness  (weight 2, scored 0-4)

     grade_records.py record-project P05 --scores 3,4,2,3,4,3
   ```

   `grade_records.py rubric all` lists every project id. `S3 - Project Portfolio.ipynb` has the brief for each of the twenty sections, with the deliverables to tick and the README entry to copy into your project repository.

2. **Build it**, in its own repository, not in this course repo. One repository per project: a single repository called "python-exercises" with twenty folders reads as coursework, twenty small repositories read as a body of work.

3. **Test it yourself.** The rubric gives tests a weight of 2 out of 12, which is a fifth of the mark. A test suite that passes from a clean checkout is the cheapest criterion to earn and the one that makes the other four believable.

4. **Write the README.** What it is, in the first two lines. A "How to run it" section with the exact commands. A screenshot or a short demo. This is the Documentation criterion, and it is also the part a recruiter reads first.

5. **Self-score against the rubric,** honestly, criterion by criterion, then record it:

   ```

   Six numbers, in rubric order, each between 0 and 4. The mark is the weighted total over the maximum of 48. Project levels: 90% excellent, 75% strong, 60% pass, 40% developing, below that weak.

Because the rubric is identical for every project, the average across your projects is meaningful: it is the portfolio average in the summary, and the weakest criterion across all of them is the honest answer to "what should I improve next".

## 8. Your progress log

Marks tell you where you are. The journal tells you how you got there, and it is the thing a reviewer reads.

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade_records.py log "rebuilt the OOP model, still stuck on super()"
```

```
journal entry 1 recorded, chain hash 1520a9b9cef8e5b2

A journal entry is a dated note: what you did, what you got stuck on, what you intend to try next. Add `--tag` to label it (`--tag oop`, `--tag review`). The entries are rendered in the "Progress journal" table of `S2 - Grading Records.ipynb` and in `GRADE_REPORT.md`.

**The ledger is a hash chain.** Every entry in `records.json` - attempts, project scores and journal notes alike - carries the hash of the entry before it (`prev`) and the hash of its own content (`hash`), plus the SHA-256 and the git blob id of the notebook that was graded. Edit a past entry and every hash after it stops matching, and `integrity.py verify` names the entry where the chain breaks:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\integrity.py verify
```

On a healthy system it prints:

```
========================================================================
INTEGRITY REPORT - Complete Python 3 Bootcamp test suite
========================================================================
ledger chain   : OK (25 entries, head 8c1f0a4b2d3e5f60)
suite manifest : OK (36 files)
seal           : OK
attestations   : OK (3 signed statements)
work drift     : none

RESULT: integrity verified
```

If something has been tampered with, the same command says so and names the problem - `ledger chain : BROKEN`, with a line for each entry whose hash no longer matches its content, or `suite manifest : CHANGED`, with the files that differ. It exits with status 1 when it finds a problem, so it belongs in a CI job.

So your history cannot be quietly rewritten. Not by you, and not by anyone else with access to the folder.

**The seal.** `integrity.py seal` signs the current state - the chain head, the hash of the manifest of every file in the suite, and the list of attestations - with an Ed25519 key:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\integrity.py seal
```

```
sealed: chain head 8c1f0a4b2d3e5f60 (25 entries), manifest 4b2e91c7d0a3f685
seal   : ...\20-Testing-Syllabus\SEAL.json
anchor : publish this hash somewhere you do not control -
         8c1f0a4b2d3e5f60...
```

The private key lives outside the repository, at `%USERPROFILE%\.professor\professor_key.pem`, and must never be committed; the public key `agent.pub` is in the folder. Sealing needs that key: if you see "no agent key - run: python integrity.py keygen", generate it once with

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\integrity.py keygen
```

and back it up somewhere safe. With the key you can keep signing; without it you can still verify every seal already made.

The seal is only as strong as where you anchor it. `INTEGRITY.md` covers anchoring the hash somewhere you do not control - a GitHub attestation, an OpenTimestamps stamp, an email to yourself - which turns the chain from decoration into evidence.

## 9. The weekly loop

One hour a session, five moves. The order matters: nothing is recorded until it is verified, and nothing is committed until it is sealed.

**1. Study (25 min).** Work through the lecture notebooks for the next section in `S0 - Testing Syllabus.ipynb`. Read the code, then retype it. The most commonly missed topics are listed per module in the blueprint, so if you already know a section, revise those topics rather than re-reading everything.

**2. Practise (15 min).** Re-do the exercises that tripped you up, or work the practice artifacts: `PRACTICE - Practice Test.ipynb` for a rehearsal of a test, `PRACTICE - Practice Project - Book Tracker.ipynb` for a rehearsal of a project.

**3. Test (15 min).** Take the matching `T**` notebook, grade it, fix the failures, and log the attempt:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade_records.py run "T03 - Methods and Functions Test.ipynb" --prose 7
```

**4. Log (2 min).** One line in the journal about what you actually did:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\grade_records.py log "T03 retake: kwargs finally clicked, mutable defaults still not"
```

**5. Seal and commit (3 min).** Re-seal so the signature covers the new chain head, then commit and push:

```
C:\Users\andre\anaconda3\python.exe 20-Testing-Syllabus\integrity.py seal
git add -A
git commit -m "T03: 19/20, signed off"
git push
git tag -s v1.0-T03-pass -m "T03 signed off: 19/20"
git push origin v1.0-T03-pass
```

Two independent layers, both worth having. Git history is the record of your *work*: the answer cells as you wrote them, the diffs between attempts, the README as it evolved. The hash chain and the seal are the record of your *grades*: which attempt scored what, and that the suite you were graded against has not changed since. Each attempt stores the git blob id of the notebook as it was graded, so if you edit an answer cell afterwards, `integrity.py verify` reports the drift and `git log` shows the change.

Eight weeks of this is the plan in `S0 - Testing Syllabus.ipynb`: weeks 1 to 3 on sections 00 to 03, weeks 4 and 5 on OOP and exceptions (the two sections that gate the most later material), weeks 6 and 7 on the rest, week 8 on the milestone projects and the final exam. If you are short on time, T00, T02, T03, T05, T07 and T17 carry 50% of the section weight: do those thoroughly.

## 10. Troubleshooting

| Symptom | What it means | What to do |
| --- | --- | --- |
| `FAIL Q9: NameError: name 'square' is not defined` | The identifier in the question does not exist in your answer | Create the exact name the question gives. `square`, not `sq`. |
| A check fails in the grader but passed in Jupyter | The answer cell was never saved; the grader reads the file on disk | `Ctrl+S` in JupyterLab, then run the grader again |
| Your answers disappeared | The kernel was restarted, which clears the namespace | Save before restarting, and keep scratch work in a separate notebook |
| `ModuleNotFoundError: No module named 'requests'` | An optional package from the course is not installed | `pip install requests beautifulsoup4 pillow pypdf openpyxl` in the Anaconda prompt |
| `SKIP: pypdf not installed` printed by a check | The optional library is absent and the check skipped itself | Not a failure. The affected sections still mark cleanly without the extra packages |
| `python` is not recognised | `python` is not on the path | Use the full path: `C:\Users\andre\anaconda3\python.exe` |
| `no notebooks found` from `grade.py` | Wrong working directory | Run from the repository root, or pass the file name explicitly |
| `unknown test: PRACTICE - Practice Test.ipynb` | The practice notebooks are not module tests, so the ledger has no record for them | Grade them with `grade.py "PRACTICE - Practice Test.ipynb"`; do not log them with `grade_records.py run` |
| `integrity.py verify` says `suite manifest : CHANGED` and names a file | A file in `20-Testing-Syllabus/` was modified, added or removed after the manifest was written | If you changed it on purpose, re-hash the suite: `integrity.py manifest`, then `integrity.py seal` |
| `integrity.py verify` says `seal : PROBLEM` | The seal is stale or was replaced | Re-seal: `integrity.py seal`. If it says the sealed chain head is not in the ledger, the seal was overwritten |
| `integrity.py seal` says `no agent key` | The private key is not on this machine | `integrity.py keygen` once, and back the key up outside the repository |
| `integrity.py verify` reports work drift | The notebook changed after it was graded | Expected if you edited your answers after logging; check `git log` to confirm the change was yours |
| A test hangs | Something in your answer is waiting for input | Interrupt the cell (`Kernel > Interrupt Kernel`) and look for `input()` |
