## Q2. Multiple choice

Errors that are detected while a statement is being executed are called:

A) Syntax errors
B) Exceptions
C) Warnings
D) Comments

<details>
<summary>Solution</summary>

**Correct answer: B.** Exceptions. Even if a statement is syntactically correct it can
still fail when it is executed; errors detected during execution are called exceptions,
and they are not unconditionally fatal because you can catch them with `try` and
`except`.

</details>

## Q3. Multiple choice

Which block of a `try` statement always runs, no matter what happens?

A) `except`
B) `else`
C) `finally`
D) `break`

<details>
<summary>Solution</summary>

**Correct answer: C.** `finally`. The `finally:` block is always executed, even if an
exception was raised in the `try` block, which makes it the right place for cleanup
such as closing a file. The `else` block runs only when no exception was raised.

</details>

## Q4. Multiple choice

When does the `else` block of a `try` / `except` statement run?

A) Only when the `try` block raised an exception
B) Only when the `try` block finished without raising an exception
C) Always, after the `finally` block
D) Only when the `except` block handled the exception

<details>
<summary>Solution</summary>

**Correct answer: B.** Only when the `try` block finished without raising an exception.
`else` is the "no error happened" branch, while `except` handles the error branches
and `finally` always runs afterwards.

</details>

## Q5. Conceptual

What is the difference between `except:` and `except IOError:`? Which one should you
prefer and why?

<details>
<summary>Solution</summary>

**Answer:** `except:` catches every exception, so the program keeps running no matter what
went wrong, but you also lose the information about what failed. `except IOError:`
catches one specific exception only, which is the problem you actually expected. The
course notes prefer the specific form: catching only the exception you can handle
means that real bugs still surface, and the `except` block can print a meaningful
message such as "Error: Could not find file or read data".

</details>

## Q6. Conceptual

What does the `raise` statement do, and how do you raise an error of your own?

<details>
<summary>Solution</summary>

**Answer:** `raise` raises an exception. You can re-raise one of the built-in
exceptions, for example `raise ValueError("Insufficient funds")`, or define your own
exception class by inheriting from `Exception` and raise instances of it, for example
`class InvalidInputError(Exception): pass` followed by
`raise InvalidInputError("age must be a whole number")`. The course also uses the
abstract method pattern where a base class raises `NotImplementedError` to force
subclasses to provide the real implementation.

</details>

## Q9. Coding (2 points)

Define:

* `safe_divide(a, b)` returning `a / b` as a float; when `b` is zero it must raise a
  `ValueError` with the message `'division by zero'` instead of letting a
  `ZeroDivisionError` escape

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def safe_divide(a, b):
    if b == 0:
        raise ValueError("division by zero")
    return a / b
```

**Why:** Checking the argument and raising your own `ValueError` gives the caller a clear message instead of the raw `ZeroDivisionError`.

</details>

````python
assert safe_divide(6, 3) == 2.0
assert safe_divide(7, 2) == 3.5
try:
    safe_divide(1, 0)
except ValueError as error:
    assert str(error) == 'division by zero'
else:
    raise AssertionError("safe_divide(1, 0) should raise ValueError")
````

  If the conversion succeeds it appends the string `'ok'` to the `log` list and
  returns the integer; if it fails it appends the string `'error'` to the `log` list
  and returns `None`. In both cases it must append the string `'finally'` to the `log`
  list last.

Use `try` / `except` / `else` / `finally` to do it.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def convert_with_log(text, log):
    try:
        value = int(text)
    except ValueError:
        log.append("error")
        return None
    else:
        log.append("ok")
        return value
    finally:
        log.append("finally")
```

**Why:** `else` runs only when no exception was raised and `finally` runs in both cases, so the log always ends with 'finally'.

</details>

````python
log = []
assert convert_with_log('42', log) == 42
assert log == ['ok', 'finally']

log2 = []
assert convert_with_log('abc', log2) is None
assert log2 == ['error', 'finally']
````

## Q11. Coding (2 points)

Define your own exception class **`InvalidInputError`** that inherits from `Exception`,
and define:

* `parse_age(text)` that converts `text` to an integer age and returns it. It must
  raise `InvalidInputError` when the text is not a whole number and also when the
  resulting age is negative. The message should mention the bad value.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
class InvalidInputError(Exception):
    pass


def parse_age(text):
    try:
        age = int(text)
    except (TypeError, ValueError):
        raise InvalidInputError("age must be a whole number: %s" % (text,))
    if age < 0:
        raise InvalidInputError("age cannot be negative: %s" % (text,))
    return age
```

**Why:** Subclassing `Exception` lets you catch your own error separately from the built-in ones, and raising it replaces the less helpful `ValueError`.

</details>

````python
assert issubclass(InvalidInputError, Exception)
assert parse_age('30') == 30
for bad in ['abc', '-5']:
    try:
        parse_age(bad)
    except InvalidInputError:
        pass
    else:
        raise AssertionError("parse_age(%r) should raise InvalidInputError" % (bad,))
````

## Q12. Coding (2 points)

Define:

* `parse_number(text)` that returns `int(text)` for a string; it must raise a
  `TypeError` when `text` is not a string and a `ValueError` when the string does not
  hold a number
* `parse_or_none(text)` that calls `parse_number` and returns the number, but returns
  `None` instead of raising when either a `TypeError` or a `ValueError` happens

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def parse_number(text):
    if not isinstance(text, str):
        raise TypeError("text must be a string")
    return int(text)


def parse_or_none(text):
    try:
        return parse_number(text)
    except TypeError:
        return None
    except ValueError:
        return None
```

**Why:** Catching the specific exceptions you know how to handle lets `parse_or_none` be a safe wrapper, while `parse_number` still reports the problem.

</details>

````python
assert parse_number('12') == 12
try:
    parse_number(12)
except TypeError:
    pass
else:
    raise AssertionError("parse_number(12) should raise TypeError")
try:
    parse_number('abc')
except ValueError:
    pass
else:
    raise AssertionError("parse_number('abc') should raise ValueError")
assert parse_or_none('abc') is None
assert parse_or_none(12) is None
assert parse_or_none('7') == 7
````

## Q14. Coding (2 points)

Write a small unit test suite in the style of the course's `test_cap.py`:

* define `cap_text(text)` returning `text.title()`
* define a test class **`TestCapTest`** that inherits from `unittest.TestCase`
* give the class a `setUp(self)` method that stores the string `'python'` in
  `self.text`
* add a test method `test_one_word` that checks `cap_text(self.text)` equals
  `'Python'`
* add a test method `test_multiple_words` that checks `cap_text('monty python')`
  equals `'Monty Python'`

Both test methods must use `self.assertEqual(...)`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
import unittest


def cap_text(text):
    return text.title()


class TestCapTest(unittest.TestCase):

    def setUp(self):
        self.text = 'python'

    def test_one_word(self):
        result = cap_text(self.text)
        self.assertEqual(result, 'Python')

    def test_multiple_words(self):
        result = cap_text('monty python')
        self.assertEqual(result, 'Monty Python')
```

**Why:** `unittest` runs every method whose name starts with `test` and calls `setUp` before each one, so every test starts from a clean state.

</details>

````python
import io
import unittest

assert hasattr(TestCapTest, 'setUp')
suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestCapTest)
result = unittest.TextTestRunner(stream=io.StringIO(), verbosity=0).run(suite)
assert result.testsRun == 2, 'expected two test methods, ran %s' % result.testsRun
assert result.wasSuccessful()
assert cap_text('monty python') == 'Monty Python'
````
