**Q1 (multiple choice, 1 point)** - What is the difference between a method and a function?

A. A method is a function attached to an object; it is called on the object (like `object.method()`) and receives that object as its first argument, conventionally called `self`.

B. A method must always be defined inside a `class`, while a function can never be defined inside a `class`.

C. A function can only return a value, while a method can never return a value.

D. There is no difference at all - the two words are interchangeable.

<details>
<summary>Answer</summary>

**A. A method is a function attached to an object; it is called on the object and receives that object as its first argument, conventionally called `self`.**

**Why:** Methods are functions built into objects: `lst.append(4)` calls the `append` method that belongs to the list, and the list itself is passed in as that first argument (you never see `self`, Python supplies it for you). A plain function such as `len(lst)` is called with its arguments only. Methods can return values just like functions.

</details>

**Q2 (multiple choice, 1 point)** - Consider the function `def f(a, b): print(a + b)`. What is the value of `x` after this line runs: `x = f(1, 2)`?

A. `3`

B. `None`

C. `0`

D. The line raises an error.

<details>
<summary>Answer</summary>

**B. `None`**

**Why:** `print()` displays its output but does not give anything back, so the function returns `None` and `x` is bound to `None`. If you want to store the result you have to use the `return` keyword: `def f(a, b): return a + b`, and then `x = f(1, 2)` gives `x` the value `3`.

</details>

**Q3 (multiple choice, 1 point)** - Given the definition `def f(a, b=2): return a + b`, which of the following calls is **not** valid Python?

A. `f(1, 3)`

B. `f(a=1, b=3)`

C. `f(1, b=3)`

D. `f(a=1, 3)`

<details>
<summary>Answer</summary>

**D. `f(a=1, 3)`**

**Why:** A positional argument has to come before any keyword argument in a call, so `f(a=1, 3)` is rejected by Python with a `SyntaxError` ("positional argument follows keyword argument"). The other three calls are all valid: `b` may be given positionally or by keyword, and the default `b=2` is only used when `b` is not supplied at all.

</details>

**Q4 (multiple choice, 1 point)** - Inside a function, what does `**kwargs` collect?

A. All of the extra positional arguments, as a list.

B. All of the extra positional arguments, as a tuple.

C. All of the keyword arguments, as a dictionary.

D. All of the keyword arguments, as a set.

<details>
<summary>Answer</summary>

**C. All of the keyword arguments, as a dictionary.**

**Why:** A parameter written as `**kwargs` gathers the keyword arguments into a dictionary, so `kwargs['fruit']` gives the value that was passed for the keyword `fruit`. The single star `*args` is the one that gathers the extra *positional* arguments into a tuple, and the word `kwargs` itself is only a convention - any name would work.

</details>

**Q5 (conceptual, 1 point)** - Predict the output of this code:

```text
x = 25

def printer():
    x = 50
    return x

print(x)
print(printer())
```

<details>
<summary>Answer</summary>

```text
25
50
```

**Why:** Assigning a name inside a function creates a **local** name, so the `x = 50` line has no effect on the global `x`, which is still 25. Inside `printer()` the local `x` is 50, and that is the value the function returns.

</details>

**Q6 (conceptual, 1 point)** - Predict the output of this code:

```text
count = 0

def bump():
    global count
    count += 1

bump()
bump()

print(count)
```

<details>
<summary>Answer</summary>

```text
2
```

**Why:** The `global count` statement tells Python that `count` inside `bump()` is the same name as the module level `count` and not a new local one, so the two calls really do change it from 0 to 1 to 2. Without the `global` statement, `count += 1` would create a local variable and the global one would stay at 0.

</details>

**Q7 (conceptual, 1 point)** - Predict the output of this code:

```text
nums = [1, 2, 3, 4, 5]

print(list(map(lambda n: n ** 2, nums)))
print(list(filter(lambda n: n % 2 == 1, nums)))
```

<details>
<summary>Answer</summary>

```text
[1, 4, 9, 16, 25]
[1, 3, 5]
```

**Why:** `map()` applies the function to every item of the iterable, so every number is squared. `filter()` keeps only the items for which the function returns `True`, and `n % 2 == 1` is true for the odd numbers. A lambda body is a single expression, which is why it can be used here but not where you would need several statements.

</details>

**Q9 (coding exercise, 2 points)** - Write a function `greet(name, greeting='Hello')` that returns the string `'<greeting>, <name>!'`.

For example `greet('Jose')` returns `'Hello, Jose!'` and `greet('Ana', greeting='Hi')` returns `'Hi, Ana!'`. The second parameter has a default value and may also be passed by keyword.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def greet(name, greeting='Hello'):
    return f'{greeting}, {name}!'
```

**Why:** A default value in the parameter list makes the second argument optional, and the caller can still pass it positionally or by keyword. The function returns the result instead of printing it, so the caller can store it.

</details>

````python
assert greet('Jose') == 'Hello, Jose!'
assert greet('Ana', greeting='Hi') == 'Hi, Ana!'
assert greet(name='Bob', greeting='Hey') == 'Hey, Bob!'
````

**Q10 (coding exercise, 2 points)** - Write a function `number_stats(numbers)` that takes a list of numbers and returns a **tuple** with three values: the smallest number, the largest number and the average of the numbers.

The function should return the tuple, not print it.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def number_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)
```

**Why:** Returning several values with commas creates a tuple, which the caller can unpack: `low, high, average = number_stats([3, 1, 4])`.

</details>

````python
result = number_stats([3, 1, 4, 1, 5])
assert isinstance(result, tuple)
low, high, average = result
assert low == 1
assert high == 5
assert abs(average - 2.8) < 1e-9
````

Calling it with no arguments at all must return `0`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def sum_all(*args):
```

**Q12 (coding exercise, 2 points)** - Write a function `keyword_keys(**kwargs)` that accepts any number of keyword arguments and returns a **sorted list** with the names of the keyword arguments that were passed.

Calling it with no arguments must return an empty list.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def keyword_keys(**kwargs):
    return sorted(kwargs.keys())
```

**Why:** `**kwargs` gathers the keyword arguments into a dictionary, so `kwargs.keys()` gives the argument names and `sorted()` puts them in alphabetical order.

</details>

````python
assert keyword_keys(b=1, a=2) == ['a', 'b']
assert keyword_keys(x=1, y=2, z=3) == ['x', 'y', 'z']
assert keyword_keys() == []
````

**Q13 (coding exercise, 2 points)** - Write a function `double_all(nums)` that takes a list of numbers and returns a **list** with every number doubled.

Use `map()` with a `lambda` expression to do the work.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def double_all(nums):
    return list(map(lambda n: n * 2, nums))
```

**Why:** `map()` applies the lambda to every item of the list, and casting the result with `list()` gives a list back instead of a map object.

</details>

````python
assert double_all([1, 2, 3]) == [2, 4, 6]
assert double_all([]) == []
assert double_all([-1, 0, 5]) == [-2, 0, 10]
````

**Q14 (coding exercise, 2 points)** - Create a global variable `counter = 0` and a function `increment_counter()` that increases that global variable by 1 and returns its new value.

Inside the function you must use the `global` statement.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
counter = 0

def increment_counter():
    global counter
    counter += 1
    return counter
```

**Why:** Without `global counter` the line `counter += 1` would create a local variable and the global one would never change. The `global` statement maps the name to the module level variable.

</details>

````python
assert counter == 0
assert increment_counter() == 1
assert increment_counter() == 2
assert counter == 2
````
