
</details>

**Q6 (coding exercise, 2 points)** - List comprehensions.

Start from `numbers = [1, 2, 3, 4, 5]` and use a **list comprehension** for each of these three variables:

- `squares`: the square of every number in `numbers`
- `evens`: only the even numbers from `numbers`, in order
- `labels`: every number in `numbers` turned into a string, for example `1` becomes `'1'`

A list comprehension has the shape `[value for item in list if condition]`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
numbers = [1, 2, 3, 4, 5]
squares = [n * n for n in numbers]
evens = [n for n in numbers if n % 2 == 0]
labels = [str(n) for n in numbers]
```

**Why:** The expression before `for` is the value that goes into the new list. The `if` part is optional and keeps only the items that satisfy it, so `n % 2 == 0` selects the even numbers. `str(n)` converts a number to its text form.

</details>

````python
assert squares == [1, 4, 9, 16, 25]
assert evens == [2, 4]
assert labels == ['1', '2', '3', '4', '5']
````

**Q7 (coding exercise, 2 points)** - Functions with default arguments.

Define the function `greet(name, greeting='Hello')`, which returns `greeting + ', ' + name`. Then create:

- `default_greeting`: the result of `greet('Ada')`
- `custom_greeting`: the result of `greet('Ada', 'Hi')`

Define a second function `repeat(text, times=2)`, which returns `text` repeated `times` times. Then create:

- `twice`: the result of `repeat('ha')`
- `four_times`: the result of `repeat('ha', 4)`

A parameter with a default value lets the caller leave that argument out.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def greet(name, greeting='Hello'):
    return greeting + ', ' + name


def repeat(text, times=2):
    return text * times


default_greeting = greet('Ada')
custom_greeting = greet('Ada', 'Hi')
twice = repeat('ha')
four_times = repeat('ha', 4)
```

**Why:** A default value is written in the function signature as `greeting='Hello'`. When the caller omits that argument the default is used; when the caller passes one, the default is ignored. `text * times` repeats a string, so `'ha' * 2` is `'haha'`.

</details>

````python
assert default_greeting == 'Hello, Ada'
assert custom_greeting == 'Hi, Ada'
assert twice == 'haha'
assert four_times == 'hahahaha'
````

**Q8 (coding exercise, 2 points)** - Counting words with a dictionary.

Write a function `word_counts(text)` that returns a dictionary mapping each word to the number of times it appears in `text`. Words are separated by spaces, and case is ignored, so `'Cat'` and `'cat'` count as the same word.

Then create:

- `counts`: the result of `word_counts('the cat sat on the mat the end')`
- `unique_words`: the number of distinct words in that text, so `len(counts)`

**If your first attempt fails, read the `AssertionError` carefully.** The message shows both the value the check expected and the value your code actually produced, and the difference between those two values is your bug. Fix your code, not the check.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def word_counts(text):
    counts = {}
    for word in text.lower().split():
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
    return counts


counts = word_counts('the cat sat on the mat the end')
unique_words = len(counts)
```

**Why:** `text.lower()` makes the count case insensitive and `split()` splits the text on spaces. For each word, the dictionary either gets a new key with the value 1, or an existing key is bumped up by one. `len(counts)` is the number of distinct words.

</details>

````python
assert counts == {'the': 3, 'cat': 1, 'sat': 1, 'on': 1, 'mat': 1, 'end': 1}
assert unique_words == 6
assert counts['the'] == 3 and counts.get('dog') is None
````
