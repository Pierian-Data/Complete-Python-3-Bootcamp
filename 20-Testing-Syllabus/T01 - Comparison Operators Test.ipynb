**Q1 (multiple choice, 1 point)** - What is the difference between `=` and `==` in Python?

A. `=` compares two values and returns `True` or `False`; `==` assigns a value to a name.

B. `=` assigns a value to a name; `==` compares two values and returns `True` or `False`.

C. They are interchangeable - `=` also returns `True` or `False`.

D. `=` is only used inside a function and `==` is only used at the top level of a script.

<details>
<summary>Answer</summary>

**B. `=` assigns a value to a name; `==` compares two values and returns `True` or `False`.**

**Why:** `=` is the assignment operator: it binds the value on the right to the name on the left, so `x = 5` stores `5` in `x`. `==` is a comparison operator: it asks whether the two operands have the same value and evaluates to the boolean `True` or `False`, so `x == 5` is `True`.

</details>

**Q2 (multiple choice, 1 point)** - Which of the following expressions evaluates to `False`?

A. `2 == 2.0`

B. `2 is 2.0`

C. `2 != 3`

D. `2 >= 2`

<details>
<summary>Answer</summary>

**B. `2 is 2.0`**

**Why:** The four results are `True`, `False`, `True`, `True`. `==` compares *values*, and `2` and `2.0` have the same value, so `2 == 2.0` is `True`. `is` compares *identity* - it asks whether both names point at the same object - and the `int` 2 and the `float` 2.0 are two different objects, so `2 is 2.0` is `False`. `2 != 3` is `True` because the values differ, and `2 >= 2` is `True` because the values are equal.

</details>

**Q3 (multiple choice, 1 point)** - The chained comparison `1 < 2 < 3` is exactly equivalent to which of the following expressions?

A. `1 < 2 and 2 < 3`

B. `1 < 2 or 2 < 3`

C. `(1 < 2) < 3`

D. `1 < (2 < 3)`

<details>
<summary>Answer</summary>

**A. `1 < 2 and 2 < 3`**

**Why:** A chained comparison is shorthand for a Boolean expression built with `and`: `1 < 2 < 3` checks `1 < 2` **and** `2 < 3`, exactly like `1 < 2 and 2 < 3`. Option B would also be `True` here, but it is a weaker test - it only needs one side to hold. Options C and D are not the same test at all: C compares the boolean `True` with `3`, and D compares `1` with the boolean `True`.

</details>

**Q4 (multiple choice, 1 point)** - What happens in Python when you evaluate `1 < 'a'`?

A. It returns `False`.

B. It returns `True`.

C. It raises a `TypeError`.

D. It returns `None`.

<details>
<summary>Answer</summary>

**C. It raises a `TypeError`.**

**Why:** An ordering comparison such as `<`, `>`, `<=` or `>=` needs two operands that can be ordered against each other. An `int` and a `str` cannot be ordered, so Python raises a `TypeError`. Equality is more forgiving: `1 == 'a'` simply returns `False`, because the two values are of different types.

</details>

**Q6 (conceptual, 1 point)** - Predict the output of this code:

```text
print(not 1 == 2)
print(1 == 2 and 2 < 3)
print(1 == 2 or 2 < 3)
```

<details>
<summary>Answer</summary>

```text
True
False
True
```

**Why:** Comparisons bind more tightly than `not`, so `not 1 == 2` means `not (1 == 2)`, which is `True`. `and` needs **both** sides to be true (`1 == 2` is false, so the result is `False`), while `or` needs only **one** side to be true (`2 < 3` is true, so the result is `True`).

</details>

**Q7 (conceptual, 1 point)** - Predict the output of this code:

```text
print(10 > 5 > 3)
print(10 > 5 > 7)
print(1 < 2 < 3 < 4)
```

<details>
<summary>Answer</summary>

```text
True
False
True
```

**Why:** Each chain is a series of `and`-ed comparisons that share the middle value: `10 > 5 > 3` is `10 > 5 and 5 > 3`, and both halves hold, so it is `True`. `10 > 5 > 7` is `10 > 5 and 5 > 7`, and the second half fails, so the whole chain is `False`. A chain is `True` only when every link in it holds.

</details>

**Q8 (conceptual, 1 point)** - Predict the output of this code:

```text
print([1, 2] == [1, 2])
print([1, 2] is [1, 2])
```

<details>
<summary>Answer</summary>

```text
True
False
```

**Why:** `==` asks whether the two lists hold equal *values*, and both lists contain 1 and 2 in the same order, so the result is `True`. `is` asks whether both names refer to the *same object*, and each list literal creates a brand new list, so the result is `False`.

</details>

**Q10 (coding exercise, 2 points)** - Write a function `compare(a, b)` that compares two numbers and returns the **string** `'equal'`, `'greater'` or `'less'`:

- return `'equal'` if `a == b`
- return `'greater'` if `a > b`
- return `'less'` in every other case

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def compare(a, b):
    if a == b:
        return 'equal'
    elif a > b:
        return 'greater'
    else:
        return 'less'
```

**Why:** An `if` / `elif` / `else` chain checks the conditions in order, so the `else` branch is only reached when the two values are neither equal nor in the order `a > b`.

</details>

````python
assert compare(5, 5) == 'equal'
assert compare(7, 3) == 'greater'
assert compare(3, 7) == 'less'
````

**Q11 (coding exercise, 2 points)** - Write a function `in_range(value, low, high)` that returns `True` when `value` lies **inside** the inclusive range from `low` to `high` (so `low` and `high` themselves are allowed), and `False` otherwise.

Write the test as a single chained comparison.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def in_range(value, low, high):
    return low <= value <= high
```

**Why:** The chained comparison `low <= value <= high` is shorthand for `low <= value and value <= high`, and because the comparisons are `<=` the endpoints are included.

</details>

````python
assert in_range(5, 1, 10) is True
assert in_range(10, 1, 10) is True
assert in_range(0, 1, 10) is False
````

**Q12 (coding exercise, 2 points)** - Write a function `is_strictly_between(value, low, high)` that returns `True` only when `value` is **strictly** between `low` and `high` (the endpoints do **not** count), and `False` otherwise.

Use a chained comparison.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def is_strictly_between(value, low, high):
    return low < value < high
```

**Why:** Using `<` on both sides of the chain excludes the endpoints, so `is_strictly_between(1, 1, 10)` is `False`.

</details>

````python
assert is_strictly_between(5, 1, 10) is True
assert is_strictly_between(1, 1, 10) is False
assert is_strictly_between(10, 1, 10) is False
````

**Q13 (coding exercise, 2 points)** - Write a function `either_is_zero(a, b)` that returns `True` when **at least one** of the two numbers is zero (so use the `or` operator), and `False` when neither of them is zero.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def either_is_zero(a, b):
    return a == 0 or b == 0
```

**Why:** `or` is true as soon as one side is true, so the function reports `True` for `(0, 5)` and for `(5, 0)`, and `False` only when both comparisons fail.

</details>

````python
assert either_is_zero(0, 5) is True
assert either_is_zero(5, 0) is True
assert either_is_zero(1, 2) is False
````

**Q14 (coding exercise, 2 points)** - Write a function `increasing(a, b, c, d)` that returns `True` when the four numbers are in strictly increasing order (`a < b < c < d`), and `False` otherwise.

Write the test as a single chained comparison.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def increasing(a, b, c, d):
    return a < b < c < d
```

**Why:** The chain `a < b < c < d` is evaluated as `a < b and b < c and c < d`, so every link has to hold for the result to be `True`.

</details>

````python
assert increasing(1, 2, 3, 4) is True
assert increasing(1, 3, 2, 4) is False
assert increasing(4, 3, 2, 1) is False
````
