## Q1. Multiple choice

Which widget property holds the current state of a widget, for example the text currently typed into a `Text` widget or the option currently chosen in a `Dropdown`?

A. `.value`
B. `.keys`
C. `.description`
D. `.children`

<details>
<summary>Solution</summary>

**Correct answer: A**

Every widget shares a common naming scheme: `.value` reads and writes the current state, `.description` is the label shown next to the control, `.keys` lists the synchronised traitlets of the widget, and `.children` is used by the container widgets (`VBox`, `HBox`, `Tab`) to hold the widgets they contain.

</details>

## Q3. Multiple choice

When you write `interact(f, x=True)`, which widget does `interact` create for the argument `x`?

A. `Checkbox`
B. `RadioButtons`
C. `Text`
D. `IntSlider`

<details>
<summary>Solution</summary>

**Correct answer: A**

`interact` turns each keyword argument into a widget based on its type. A boolean (`True` / `False`) becomes a `Checkbox`, a string becomes a `Text` widget, an integer becomes an `IntSlider`, a float becomes a `FloatSlider`, and a list or a dict becomes a `Dropdown`. D. It disables the button so it cannot be clicked again

<details>

## Q5. Conceptual

What is the difference between `interact` and `interactive`?

Explain what each one returns and how you would display the result of `interactive`.

<details>
<summary>Solution</summary>

**Reference answer**

`interact` builds the widgets and displays them immediately; it does not give you back an object you can inspect. Its return value is not useful.

`interactive` builds the same widgets but **returns** an `interactive` object instead of displaying it, so you can hold on to it. That object is a subclass of `VBox` (a container of the generated widgets plus an output widget) and it exposes:

- `.children` - the generated child widgets,
- `.kwargs` - the current values of the keyword arguments,
- `.result` - the current return value of the underlying function.

To show it, you must call `display(w)` explicitly, for example `w = interactive(f, a=10, b=20)` followed by `display(w)`. Also note that with `interactive` the function's return value is not displayed automatically, so you call `IPython.display.display(...)` inside the function if you want to show a value.

</details>

## Q6. Conceptual

When you register a callback with `observe`, what arguments does the callback receive, and which keys does the change dictionary contain?

Give the signature of the handler and at least three keys of the change object.

<details>
<summary>Solution</summary>

**Reference answer**

The handler is called with a single argument, the change dictionary, so the signature is `def handler(change):`. When the notification type is `'change'`, the dictionary contains at least:

- `'name'` - the name of the trait that changed, for example `'value'`,
- `'old'` - the previous value,
- `'new'` - the new value,
- `'owner'` - the widget instance the change came from,
- `'type'` - the type of notification, `'change'` here.

The usual call looks like `slider.observe(handler, names='value')`, where `names` says which trait to watch. `observe` is the general mechanism and works on any widget and any trait; `on_click` and `on_submit` are the two special events for the `Button` and the `Text` widget.

</details>

## Q7. Conceptual

What is the difference between a widget's `layout` attribute and its `style` attribute?

Give two examples of a `layout` property and two examples of a `style` property.

<details>
<summary>Solution</summary>

**Reference answer**

`layout` exposes the CSS properties of the widget's top-level DOM element, so it controls layout and sizing. It is available on every widget, and its properties map onto CSS names with underscores replaced by dashes. Examples: `layout.width`, `layout.height`, `layout.margin`, `layout.padding`, `layout.border`, `layout.align_items`, `layout.display`.

`style` exposes the non-layout appearance of a widget, and the properties it offers depend on the widget type. Examples: for a `Button`, `style.button_color` and `style.font_weight`; for an `IntSlider`, `style.handle_color` and `style.description_width`.

Because `layout` is general, it can be copied between any two widgets of the same type (`x.layout = w.layout`), and the same is true of `style` (`b2.style = b1.style`) - but the style copy does not bring along other properties such as `description`.

</details>

## Q8. Conceptual

A `Button` offers a `button_style` argument for quick colouring. List the predefined values it accepts, and say what you would use instead if you needed an arbitrary colour.

<details>
<summary>Solution</summary>

**Reference answer**

`button_style` accepts these predefined values: `'primary'`, `'success'`, `'info'`, `'warning'`, `'danger'`, and the default empty string `''` for the ordinary button. It is passed straight to the constructor, for example `widgets.Button(description='Danger Button', button_style='danger')`.

For an arbitrary colour you use the `style` attribute instead and set the `button_color` trait, for example `b1 = widgets.Button(description='Custom color')` followed by `b1.style.button_color = 'lightgreen'`. The list of style traits available for a widget can be inspected with `b1.style.keys`.

</details>

## Q9. Coding

Write a function **`make_click_handler(state)`** that returns a callback function **`handler(button)`** for use with `button.on_click(handler)`.

The callback updates the dictionary `state` **in place**:

- `state['clicks']` holds how many times the handler has been called; start it at `0` if the key is missing.
- `state['log']` holds the labels of the buttons that were clicked, in order; start it as an empty list if the key is missing.

The label of a button is its `description` attribute when it has one, and `str(button)` otherwise.

The handler must **return** the `state` dictionary.

Note: registering the handler with a real widget is optional. If `ipywidgets` is installed you may add `widgets.Button(description='Save').on_click(handler)`, but the function itself must work with plain Python objects.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def make_click_handler(state):
    def handler(button):
        label = getattr(button, "description", None)
        if label is None:
            label = str(button)
        state["clicks"] = state.get("clicks", 0) + 1
        state.setdefault("log", []).append(label)
        return state

    return handler
```

**Why:** A closure keeps the state dictionary in scope without a global variable. `state.get` and `state.setdefault` create the keys on first use, and `getattr(button, 'description', None)` reads the label only when the object actually has one.

</details>

````python
class FakeButton:
    def __init__(self, description):
        self.description = description


state = {}
handler = make_click_handler(state)
result = handler(FakeButton("Save"))

assert result is state
assert state["clicks"] == 1
assert state["log"] == ["Save"]

handler(FakeButton("Load"))
assert state["clicks"] == 2
assert state["log"] == ["Save", "Load"]
````

## Q10. Coding

Write a function **`build_label_options(items)`** that builds the list of options for a dropdown.

It takes a list of items and returns a **new list of `(label, value)` tuples**, ready to be passed as `widgets.Dropdown(options=...)`:

- `label` is the text shown in the menu: the item's position followed by a period, a space and the item, with the position starting at **1** (so the first label is `'1. Apple'`).
- `value` is the item itself.

An empty input list must produce an empty output list, and the input list must not be modified.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def build_label_options(items):
    options = []
    for position, item in enumerate(items, start=1):
        options.append(("{0}. {1}".format(position, item), item))
    return options
```

**Why:** `enumerate` with `start=1` gives the one-based position used in the label, and building a new list leaves the caller's list untouched. The same list can be handed to `widgets.Dropdown(options=...)` or to `widgets.Select(options=...)`.

</details>

````python
options = build_label_options(["Apple", "Banana"])
assert options == [("1. Apple", "Apple"), ("2. Banana", "Banana")]
assert all(isinstance(label, str) and isinstance(value, str) for label, value in options)

source = []
assert build_label_options(source) == []
assert source == []
````

## Q11. Coding

Write a function **`filter_by_choice(items, selected)`** that powers the dropdown of a small GUI.

- `items` is a list of dictionaries. Every dictionary has a `'title'` key and a `'genre'` key.
- `selected` is the value currently chosen in the dropdown. It may be `None` or an empty string, meaning "no filter".
- When there is no filter, return a **new list** containing every item (not the same object as `items`).
- Otherwise return a **new list** with only the items whose `'genre'` matches `selected`, compared **case-insensitively**, keeping the original order.

The function must not modify `items`.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def filter_by_choice(items, selected):
    if not selected:
        return list(items)

    wanted = selected.strip().lower()
    return [
        item for item in items
        if item["genre"].strip().lower() == wanted
    ]
```

**Why:** Normalising both sides with `strip()` and `lower()` makes the comparison case-insensitive and tolerant of stray spaces. `list(items)` gives the caller its own copy when no filter is active.

</details>

````python
items = [
    {"title": "Arrival", "genre": "Science Fiction"},
    {"title": "Alien", "genre": "Science Fiction"},
    {"title": "Amelie", "genre": "Drama"},
]

assert filter_by_choice(items, "science fiction") == [items[0], items[1]]
assert filter_by_choice(items, "SCIENCE FICTION") == [items[0], items[1]]
assert filter_by_choice(items, "drama") == [items[2]]

everything = filter_by_choice(items, None)
assert everything == items
assert everything is not items
assert items[0]["title"] == "Arrival"
````

## Q12. Coding

Write a function **`current_selection(checked)`** that reports which options the user has currently selected.

- `checked` is a dictionary mapping an option label to a boolean, for example the values of several `Checkbox` widgets.
- Return a **set** containing the labels whose value is truthy.
- An empty dictionary must produce an empty set.

This is the function you would call inside a widget callback to know what the user has ticked.

````python
# your answer here
````

<details>
<summary>Solution</summary>

```python
def current_selection(checked):
    return {label for label, is_checked in checked.items() if is_checked}
```

**Why:** A set comprehension over `dict.items()` keeps only the labels whose value is truthy. Returning a set makes it easy to test for membership, to compare against the previous selection, and to feed the result straight into another filter.

</details>

````python
checked = {"Drama": True, "Sci-Fi": False, "Comedy": True}
selection = current_selection(checked)

assert selection == {"Drama", "Comedy"}
assert isinstance(selection, set)
assert "Drama" in selection and "Sci-Fi" not in selection
assert current_selection({}) == set()
````
