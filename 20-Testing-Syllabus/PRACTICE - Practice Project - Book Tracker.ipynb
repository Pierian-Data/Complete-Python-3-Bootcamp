```python
import json
import os


def load_books(path):
    """Return the books stored in the JSON file at path."""
    if not os.path.exists(path):
        return []
    with open(path, "r", encoding="utf-8") as handle:
        try:
            raw = json.load(handle)
        except ValueError as error:
            raise ValueError("could not read %s: %s" % (path, error))
    books = []
    for item in raw:
        book = Book(item["title"], item["author"])
        book.read = bool(item.get("read", False))
        books.append(book)
    return books


def save_books(books, path):
    """Write the books to the JSON file at path."""
    payload = [
        {"title": book.title, "author": book.author, "read": bool(book.read)}
        for book in books
    ]
    with open(path, "w", encoding="utf-8") as handle:
        json.dump(payload, handle, indent=2)
```

**Why:** A missing file is the normal case for a first run, so it returns an empty list rather than raising. `json.dump` writes the array of objects, and the loader rebuilds `Book` objects from it, so a save and a load give you the same list back.

</details>

````python
import json
import os
import tempfile

books = [Book("Dune", "Frank Herbert"), Book("The Hobbit", "J.R.R. Tolkien")]
books[0].mark_read()

with tempfile.TemporaryDirectory() as folder:
    path = os.path.join(folder, "books.json")

    save_books(books, path)
    assert os.path.exists(path)

    with open(path, "r", encoding="utf-8") as handle:
        raw = json.load(handle)
    assert isinstance(raw, list), "the file must hold a JSON array of books"
    assert len(raw) == 2
    assert raw[0]["title"] == "Dune"
    assert raw[0]["author"] == "Frank Herbert"
    assert bool(raw[0]["read"]) is True

    loaded = load_books(path)
    assert len(loaded) == 2
    summary = [(book.title, book.author, bool(book.read)) for book in loaded]
    assert summary == [
        ("Dune", "Frank Herbert", True),
        ("The Hobbit", "J.R.R. Tolkien", False),
    ]

    missing = load_books(os.path.join(folder, "not-there.json"))
    assert missing == [], "a missing file must load as an empty list"

    broken_path = os.path.join(folder, "broken.json")
    with open(broken_path, "w", encoding="utf-8") as handle:
        handle.write("{not json")
    try:
        result = load_books(broken_path)
    except ValueError:
        result = []
    assert result == [], "unreadable JSON must not crash the loader"
````

```python
def add_book(books, title, author):
    """Append a new book to books and return it."""
    clean_title = (title or "").strip()
    if not clean_title:
        raise ValueError("a book needs a title")
    book = Book(clean_title, (author or "").strip())
    books.append(book)
    return book


def mark_as_read(books, title):
    """Mark the matching book as read. Return True if one matched."""
    wanted = (title or "").strip().lower()
    for book in books:
        if book.title.strip().lower() == wanted:
            book.read = True
            return True
    return False


def search_books(books, text):
    """Return the books whose title or author contains text."""
    needle = (text or "").strip().lower()
    if not needle:
        return []
    return [
        book for book in books
        if needle in book.title.lower() or needle in book.author.lower()
    ]
```

**Why:** `strip()` and `lower()` make the title matching forgiving without losing the exact-match rule for `read`. `search_books` checks both the title and the author, because a reader searching for a name wants either hit.

</details>

````python
books = []

add_book(books, "Dune", "Frank Herbert")
assert len(books) == 1
assert books[0].title == "Dune"
assert books[0].author == "Frank Herbert"
assert not books[0].read

try:
    add_book(books, "   ", "Nobody")
except ValueError:
    pass
else:
    raise AssertionError("add_book must reject an empty title")
assert len(books) == 1, "a rejected title must not add a book"

assert mark_as_read(books, "dune") is True
assert books[0].read
assert mark_as_read(books, "The Hobbit") is False

matches = search_books(books, "herbert")
assert len(matches) == 1
assert matches[0].title == "Dune"

assert search_books(books, "DUN") != [], "search must ignore case"
assert search_books(books, "tolkien") == []
````

```python
import os


def _data_file():
    """The JSON file the tool reads and writes."""
    return os.environ.get("BOOK_TRACKER_FILE") or os.path.join(os.getcwd(), "books.json")


def main(argv):
    """Run one command and return an exit code."""
    args = list(argv)
    if not args:
        print("usage: book_tracker.py <add|list|read|search> ...")
        return 2

    command = args[0]
    path = _data_file()

    try:
        if command == "add":
            if len(args) != 3:
                print("usage: book_tracker.py add <title> <author>")
                return 2
            books = load_books(path)
            book = add_book(books, args[1], args[2])
            save_books(books, path)
            print("added %s by %s" % (book.title, book.author))
            return 0

        if command == "list":
            books = load_books(path)
            if not books:
                print("no books yet")
            for book in books:
                print(book)
            return 0

        if command == "read":
            if len(args) != 2:
                print("usage: book_tracker.py read <title>")
                return 2
            books = load_books(path)
            if not mark_as_read(books, args[1]):
                print("no book titled %s" % args[1])
                return 1
            save_books(books, path)
            print("marked %s as read" % args[1])
            return 0

        if command == "search":
            if len(args) != 2:
                print("usage: book_tracker.py search <text>")
                return 2
            matches = search_books(load_books(path), args[1])
            if not matches:
                print("no books match %s" % args[1])
            for book in matches:
                print(book)
            return 0
    except ValueError as error:
        print("error: %s" % error)
        return 1

    print("unknown command: %s" % command)
    print("usage: book_tracker.py <add|list|read|search> ...")
    return 2
```

**Why:** Every branch returns an exit code instead of letting an exception escape, so an unknown command or a bad file becomes a message and a non-zero code. The `try` around the work turns a `ValueError` from the commands into the same kind of message.

</details>

````python
import contextlib
import io
import os
import tempfile


def run_tool(args):
    """Run main() and capture whatever it printed."""
    out = io.StringIO()
    err = io.StringIO()
    with contextlib.redirect_stdout(out), contextlib.redirect_stderr(err):
        code = main(args)
    return code, out.getvalue() + err.getvalue()


with tempfile.TemporaryDirectory() as folder:
    data_file = os.path.join(folder, "books.json")
    previous_file = os.environ.get("BOOK_TRACKER_FILE")
    previous_folder = os.getcwd()
    os.environ["BOOK_TRACKER_FILE"] = data_file
    os.chdir(folder)
    try:
        code, output = run_tool(["add", "Dune", "Frank Herbert"])
        assert code == 0, output

        saved_to = data_file if os.path.exists(data_file) else os.path.join(folder, "books.json")
        assert os.path.exists(saved_to), "the add command must save the list"

        code, output = run_tool(["list"])
        assert code == 0, output
        assert "Dune" in output

        code, output = run_tool(["read", "dune"])
        assert code == 0, output
        stored = load_books(saved_to)
        assert len(stored) == 1
        assert stored[0].read, "the read command must save the read flag"

        code, output = run_tool(["search", "herbert"])
        assert code == 0, output
        assert "Dune" in output

        code, output = run_tool(["read", "The Hobbit"])
        assert code != 0, "reading a book that is not there must fail"
        assert output.strip(), "a failed command must say why"

        code, output = run_tool(["frobnicate"])
        assert code != 0, "an unknown command must fail"
        assert output.strip(), "an unknown command must say why"
        assert "Traceback" not in output
    finally:
        os.chdir(previous_folder)
        if previous_file is None:
            os.environ.pop("BOOK_TRACKER_FILE", None)
        else:
            os.environ["BOOK_TRACKER_FILE"] = previous_file
````
